Cette cellule définit quelques fonctions de dessin réutilisées par les schémas de chaque algorithme. Elle doit être exécutée **avant** les schémas (après la cellule de données ci-dessus).

```python
# Outils de dessin pour les schémas explicatifs (Q2) — figures pédagogiques originales
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Ellipse, Circle, Polygon, Rectangle
BLEU, ORANGE, VERT, ROUGE, GRIS = '#3b5ba5', '#e07b39', '#3a9a5b', '#c0392b', '#7f8c8d'

def canevas(largeur=10, hauteur=4):
    """Zone de dessin sans axes, graduée de 0 à `largeur` et de 0 à `hauteur`."""
    fig, ax = plt.subplots(figsize=(largeur * .9, hauteur * .9))
    ax.set_xlim(0, largeur); ax.set_ylim(0, hauteur); ax.set_aspect('equal'); ax.axis('off')
    return fig, ax

def boite(ax, x, y, l, h, texte, fc='#e8f0fe', ec=BLEU, fs=9, **kw):
    ax.add_patch(FancyBboxPatch((x - l / 2, y - h / 2), l, h, boxstyle='round,pad=0.02,rounding_size=0.12', fc=fc, ec=ec, lw=1.5))
    ax.text(x, y, texte, ha='center', va='center', fontsize=fs, **kw)

def fleche(ax, p, q, texte=None, rad=0., couleur='#444', fs=8, style='-', decal=(0, .08)):
    ax.add_patch(FancyArrowPatch(p, q, arrowstyle='-|>', mutation_scale=14, lw=1.4, color=couleur,
                                 connectionstyle=f'arc3,rad={rad}', linestyle=style))
    if texte:
        ax.text((p[0] + q[0]) / 2 + decal[0], (p[1] + q[1]) / 2 + decal[1], texte, ha='center', va='bottom', fontsize=fs, color=couleur)

def ellipse_gauss(ax, moyenne, cov, n_sigma=2, **kw):
    valeurs, vecteurs = np.linalg.eigh(cov)
    angle = np.degrees(np.arctan2(vecteurs[1, -1], vecteurs[0, -1]))
    ax.add_patch(Ellipse(moyenne, 2 * n_sigma * np.sqrt(valeurs[-1]), 2 * n_sigma * np.sqrt(valeurs[0]), angle=angle, **kw))

def grille_decision(modele, xlim, ylim, pas=200):
    gx, gy = np.meshgrid(np.linspace(*xlim, pas), np.linspace(*ylim, pas))
    return gx, gy, modele(np.c_[gx.ravel(), gy.ravel()]).reshape(gx.shape)
```

## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Régression linéaire (moindres carrés ordinaires) — *Linear Regression / Ordinary Least Squares (OLS)*. **Famille :** régression supervisée. **Objectif :** prédire une variable quantitative $y$ comme combinaison linéaire de variables explicatives.
- **Intuition et étapes :** on cherche l'hyperplan qui passe « au plus près » des points au sens de l'erreur quadratique. Entrée : matrice $X$ ($n$ observations × $p$ variables) et vecteur $y$ → apprentissage : calcul des coefficients qui minimisent la somme des carrés des résidus → sortie : prédiction $\hat y = \hat\beta_0 + x^\top\hat\beta$.
- **Critère :** $\displaystyle \hat\beta=\arg\min_{\beta_0,\beta}\sum_{i=1}^{n}\big(y_i-\beta_0-x_i^\top\beta\big)^2$, avec $x_i\in\mathbb R^p$ les variables de l'observation $i$, $y_i$ sa cible, $\beta_0$ l'ordonnée à l'origine et $\beta\in\mathbb R^p$ les coefficients. Solution analytique : $\hat\beta=(X^\top X)^{-1}X^\top y$ (variables centrées).
- **Paramètres / hyperparamètres :** paramètres appris : $\beta_0,\beta$. Hyperparamètres : aucun dans la version de base (seuls restent des choix de modélisation : variables retenues, transformations).
- **Avantages :** (1) très interprétable : un coefficient est l'effet moyen d'une variable « toutes choses égales par ailleurs » ; (2) rapide, solution fermée, peu de données nécessaires.
- **Limites :** (1) suppose une relation linéaire et est sensible aux valeurs extrêmes ; (2) instable en cas de colinéarité forte entre variables ou lorsque $p$ est proche de $n$.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_regression_lineaire():
    g = np.random.default_rng(GRAINE)
    x = np.sort(g.uniform(0, 10, 25)); y = 1.5 + 0.8 * x + g.normal(0, 1.2, 25)
    b1, b0 = np.polyfit(x, y, 1)
    fig, ax = plt.subplots(figsize=(6.5, 4))
    ax.scatter(x, y, color=BLEU, zorder=3, label='observations $(x_i, y_i)$')
    xs = np.linspace(0, 10, 100)
    ax.plot(xs, b0 + b1 * xs, color=ORANGE, lw=2, label=r'droite ajustée $\hat y=\hat\beta_0+\hat\beta_1 x$')
    ax.vlines(x, y, b0 + b1 * x, color=ROUGE, lw=1, label=r'résidus $e_i=y_i-\hat y_i$')
    ax.set(xlabel='variable explicative $x$', ylabel='cible $y$', title='MCO : minimiser la somme des carrés des résidus')
    ax.legend(fontsize=8); plt.tight_layout(); plt.show()
schema_regression_lineaire()
```

**Comment lire le schéma.** Chaque point bleu est une observation ; la droite orange est le modèle ajusté ; chaque segment rouge est un résidu $e_i=y_i-\hat y_i$. Les coefficients retenus sont ceux qui rendent la somme des **carrés** de ces segments la plus petite possible. (Données illustratives générées pour le schéma.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** estimer le $\beta$ d'un titre en régressant son rendement sur celui du marché (modèle de marché / MEDAF) ; ou prévoir les ventes d'un magasin selon budget publicitaire, prix et saisonnalité (gestion / marketing).
- **Cas d'usage 2 :** estimer le prix d'un logement à partir de la surface et du nombre de pièces.
- **Données et prérequis :** cible numérique continue ; variables numériques ou encodées en indicatrices ; $n\gg p$ ; pas de colinéarité parfaite ; pour l'inférence statistique, résidus approximativement homoscédastiques et indépendants.
- **Cas peu approprié :** prédire une classe ou une probabilité (cible binaire, ex. défaut de paiement) ou une relation fortement non linéaire avec saturation : la droite prédirait des valeurs hors de $[0,1]$ ou ignorerait la courbure ; on préférera la régression logistique ou des modèles non linéaires.
- **Métriques adaptées :** MAE, RMSE, $R^2$ sur un jeu de test, complétées par l'analyse graphique des résidus.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** `Xr_ent` (210 × 5) et `yr_ent` pour l'apprentissage ; `Xr_test` (70 × 5) pour l'évaluation. Données **synthétiques** (`make_regression`, bruit = 13). **Hyperparamètres :** aucun.
- **Sortie obtenue :** MAE = **9,35** (dans l'unité de $y$) sur les 70 observations de test.
- **Ce que cela permet de conclure :** en moyenne, la prédiction s'écarte de la vraie valeur d'environ 9,4 unités. Pour un bruit gaussien d'écart-type 13, l'erreur absolue moyenne irréductible est d'environ $\sqrt{2/\pi}\times13\approx10{,}4$ : l'ordre de grandeur obtenu est cohérent avec un modèle qui a retrouvé la vraie relation (ici linéaire par construction).
- **Ce que cela ne permet pas de conclure :** une seule métrique est affichée (ni RMSE, ni $R^2$) sur un seul découpage de 70 points : aucun intervalle de confiance ; et comme les données sont synthétiques, rien n'est dit sur des données économiques réelles. ## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Régression Ridge (régularisation $L^2$, ou de Tikhonov) — *Ridge Regression*. **Famille :** régression supervisée régularisée. **Objectif :** stabiliser la régression linéaire en pénalisant la taille des coefficients.
- **Intuition et étapes :** on ajoute au critère des moindres carrés une pénalité qui « tire » les coefficients vers zéro, ce qui réduit leur variance au prix d'un léger biais. Entrée : $X$ **standardisée** et $y$ → apprentissage : minimisation du critère pénalisé → sortie : prédiction $\hat y$.
- **Critère :** $\displaystyle \hat\beta=\arg\min_{\beta_0,\beta}\sum_{i=1}^{n}\big(y_i-\beta_0-x_i^\top\beta\big)^2+\alpha\,\lVert\beta\rVert_2^2$, où $\alpha\ge0$ règle la force de la pénalité ; solution : $\hat\beta=(X^\top X+\alpha I)^{-1}X^\top y$ ($X$ centrée, $I$ matrice identité).
- **Paramètres / hyperparamètres :** paramètres : $\beta_0,\beta$ ($\beta_0$ n'est pas pénalisé). Hyperparamètre : $\alpha$ (`alpha=1.0` dans le code), à choisir par validation croisée. La standardisation est indispensable, car la pénalité dépend de l'échelle des variables.
- **Avantages :** (1) stable en présence de variables corrélées ; (2) solution analytique, calcul rapide, fonctionne même si $p>n$.
- **Limites :** (1) ne met aucun coefficient exactement à zéro : pas de sélection de variables ; (2) introduit un biais et exige de régler $\alpha$.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_ridge():
    A = np.array([[1.0, .6], [.6, 1.4]]); b = np.array([2.0, 1.2])          # A : courbure de la perte, b : minimum MCO
    perte = lambda B: np.einsum('...i,ij,...j->...', B - b, A, B - b)
    gx, gy = np.meshgrid(np.linspace(-1.6, 3.4, 300), np.linspace(-1.6, 3.4, 300))
    P = perte(np.dstack([gx, gy]))
    br = np.linalg.solve(A + 1.5 * np.eye(2), A @ b); r = np.linalg.norm(br)  # solution ridge (alpha = 1,5)
    fig, ax = plt.subplots(figsize=(5.8, 5.4))
    ax.contour(gx, gy, P, levels=sorted([.3, 1.0, 2.2, 3.6, perte(br)]), colors=BLEU, linewidths=1)
    ax.add_patch(Circle((0, 0), r, fc='#f6c9a8', ec=ORANGE, alpha=.55, lw=2))
    ax.plot(*b, 'o', color=BLEU, ms=8); ax.annotate(r'$\hat\beta^{MCO}$', b, (b[0] + .1, b[1] + .15), color=BLEU)
    ax.plot(*br, 'o', color=ROUGE, ms=8); ax.annotate(r'$\hat\beta^{ridge}$', br, (br[0] + .15, br[1] - .35), color=ROUGE)
    ax.axhline(0, color='k', lw=.5); ax.axvline(0, color='k', lw=.5)
    ax.set(xlabel=r'$\beta_1$', ylabel=r'$\beta_2$', aspect='equal', title='Ridge : ellipses de perte et boule L2 (disque orange)')
    plt.tight_layout(); plt.show()
schema_ridge()
```

**Comment lire le schéma.** Les ellipses bleues sont les courbes de niveau de la perte quadratique, centrées sur la solution MCO. Le disque orange est la région autorisée par la pénalité $L^2$ ($\lVert\beta\rVert_2\le r$). La solution Ridge (point rouge) est le point du disque le plus proche du minimum MCO : elle est **rapprochée de l'origine** mais, le disque n'ayant pas de coin, aucune coordonnée n'est exactement nulle. (Valeurs illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** modéliser un indicateur de risque ou un rendement à partir de nombreux indicateurs macro-économiques fortement corrélés (inflation, taux, PIB, indices de confiance).
- **Cas d'usage 2 :** spectroscopie ou génomique, où le nombre de variables est élevé et les variables sont corrélées.
- **Données et prérequis :** cible continue ; variables numériques **standardisées** ; colinéarité ou $p$ proche de $n$ ; choix de $\alpha$ par validation croisée.
- **Cas peu approprié :** lorsqu'on veut une sélection parcimonieuse de variables (préférer Lasso) ; ou lorsque les variables sont peu nombreuses et indépendantes : la pénalité n'apporte alors presque rien.
- **Métriques adaptées :** MAE, RMSE, $R^2$ en validation croisée (pour choisir $\alpha$), et courbe du chemin de régularisation des coefficients.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données synthétiques que l'exemple 1. **Hyperparamètres :** `alpha=1.0` (non réglé) ; `StandardScaler` placé dans un `pipeline`, donc ajusté sur l'entraînement uniquement.
- **Sortie obtenue :** MAE = **9,37**, contre 9,35 pour la régression linéaire.
- **Ce que cela permet de conclure :** sur ces données, la pénalité change à peine la performance (écart de l'ordre de 0,015). C'est attendu : $n=210$ est très supérieur à $p=5$ et rien n'indique de colinéarité, donc la régularisation n'a presque rien à corriger.
- **Ce que cela ne permet pas de conclure :** que Ridge serait « moins bon » : l'écart est négligeable et non testé statistiquement. $\alpha$ n'a pas été réglé (aucune validation croisée), et les coefficients ne sont pas affichés : on ne voit pas leur rétrécissement.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Lasso (régularisation $L^1$) — *Least Absolute Shrinkage and Selection Operator*. **Famille :** régression supervisée régularisée. **Objectif :** prédire $y$ tout en **sélectionnant automatiquement** les variables utiles.
- **Intuition et étapes :** la pénalité sur la somme des valeurs absolues des coefficients en force certains à être exactement nuls. Entrée : $X$ standardisée, $y$ → apprentissage par descente de coordonnées (pas de formule fermée) → sortie : prédiction avec un modèle parcimonieux.
- **Critère (convention scikit-learn) :** $\displaystyle \hat\beta=\arg\min_{\beta_0,\beta}\ \frac1{2n}\sum_{i=1}^{n}\big(y_i-\beta_0-x_i^\top\beta\big)^2+\alpha\,\lVert\beta\rVert_1$, avec $\lVert\beta\rVert_1=\sum_j|\beta_j|$ et $\alpha\ge0$ la force de la pénalité.
- **Paramètres / hyperparamètres :** paramètres : $\beta_0,\beta$. Hyperparamètres : $\alpha$ (`alpha=.1`), `max_iter` (critère d'arrêt de l'optimisation, ici 10 000).
- **Avantages :** (1) sélection de variables intégrée : modèle parcimonieux et lisible ; (2) utilisable quand $p>n$.
- **Limites :** (1) parmi des variables corrélées, il en retient une de façon plutôt arbitraire et instable ; (2) biaise (rétrécit) aussi les coefficients conservés, et exige de régler $\alpha$.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_lasso():
    A = np.array([[1.0, .3], [.3, 1.0]]); b = np.array([2.0, .6]); t = 1.2    # t : budget |b1|+|b2| <= t
    perte = lambda B: np.einsum('...i,ij,...j->...', B - b, A, B - b)
    s = np.linspace(-t, t, 4001)
    bord = np.vstack([np.c_[s, t - abs(s)], np.c_[s, -(t - abs(s))]])         # frontière du losange L1
    bl = bord[np.argmin(perte(bord))]
    gx, gy = np.meshgrid(np.linspace(-1.6, 3.2, 300), np.linspace(-1.6, 3.2, 300))
    fig, ax = plt.subplots(figsize=(5.8, 5.4))
    ax.contour(gx, gy, perte(np.dstack([gx, gy])), levels=sorted([.3, 1.0, 2.2, 3.6, perte(bl)]), colors=BLEU, linewidths=1)
    ax.add_patch(Polygon([[t, 0], [0, t], [-t, 0], [0, -t]], fc='#f6c9a8', ec=ORANGE, alpha=.55, lw=2))
    ax.plot(*b, 'o', color=BLEU, ms=8); ax.annotate(r'$\hat\beta^{MCO}$', b, (b[0] + .1, b[1] + .15), color=BLEU)
    ax.plot(*bl, 'o', color=ROUGE, ms=8); ax.annotate(r'$\hat\beta^{lasso}$ (coin : $\beta_2=0$)', bl, (bl[0] - .2, bl[1] - .5), color=ROUGE)
    ax.axhline(0, color='k', lw=.5); ax.axvline(0, color='k', lw=.5)
    ax.set(xlabel=r'$\beta_1$', ylabel=r'$\beta_2$', aspect='equal', title='Lasso : le losange L1 a des coins sur les axes')
    plt.tight_layout(); plt.show()
schema_lasso()
```

**Comment lire le schéma.** Même principe que pour Ridge, mais la région autorisée (losange orange) correspond à $|\beta_1|+|\beta_2|\le t$ et possède des **coins sur les axes**. Les ellipses de perte touchent le plus souvent le losange en un coin : la solution (point rouge) a alors une coordonnée **exactement nulle** ($\beta_2=0$), ce qui explique la sélection de variables. (Valeurs illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance / marketing) :** parmi des dizaines de ratios financiers ou de canaux marketing, retenir ceux qui expliquent réellement la perte attendue ou les ventes.
- **Cas d'usage 2 :** biostatistique : sélectionner quelques gènes ou biomarqueurs parmi des milliers.
- **Données et prérequis :** cible continue ; variables standardisées ; hypothèse de parcimonie (peu de variables réellement utiles) ; $\alpha$ choisi par validation croisée.
- **Cas peu approprié :** groupes de variables corrélées toutes pertinentes (le Lasso en écarte arbitrairement une partie : préférer Elastic Net ou Ridge) ; ou quand toutes les variables sont informatives, comme dans cet exemple.
- **Métriques adaptées :** RMSE, MAE, $R^2$ en validation croisée ; nombre de coefficients non nuls ; stabilité de la sélection entre échantillons.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données que l'exemple 1 (210 observations d'entraînement). **Hyperparamètres :** `alpha=.1`, `max_iter=10000`, standardisation dans le `pipeline`.
- **Sortie obtenue :** $R^2$ = **0,986** sur le jeu de test.
- **Ce que cela permet de conclure :** environ 98,6 % de la variance de $y$ est expliquée sur les 70 observations de test.
- **Ce que cela ne permet pas de conclure :** (1) ce $R^2$ n'est **pas comparable directement** aux MAE des exemples 1, 2 et 4 (métrique différente) ; (2) la cellule n'affiche pas les coefficients, donc on ne voit pas la sélection. En complément, nous avons vérifié que les 5 coefficients restent non nuls : `make_regression` rend ici les 5 variables informatives, donc le Lasso n'a aucune variable inutile à éliminer et son effet se limite à un léger rétrécissement ; (3) $\alpha$ n'est pas réglé.

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Régression polynomiale — *Polynomial Regression*. **Famille :** régression supervisée. **Objectif :** modéliser une relation courbe en gardant un modèle linéaire **en ses paramètres**.
- **Intuition et étapes :** on enrichit les variables par leurs puissances et leurs produits, puis on applique une régression linéaire (ici pénalisée par Ridge) sur ces nouvelles variables. Entrée : $X$ → construction des termes polynomiaux → ajustement → prédiction.
- **Critère (degré 2) :** $\displaystyle \hat y=\beta_0+\sum_{j=1}^{p}\beta_jx_j+\sum_{1\le j\le k\le p}\beta_{jk}\,x_jx_k$, où $x_j$ est la $j$-ième variable et $\beta_{jk}$ le coefficient du terme croisé ; le nombre de termes croît comme $\binom{p+d}{d}-1$ (20 termes pour $p=5$, $d=2$).
- **Paramètres / hyperparamètres :** paramètres : les $\beta$. Hyperparamètres : le **degré** $d$ (`degree=2`) et la pénalité `alpha=10` du Ridge.
- **Avantages :** (1) capture courbure et interactions tout en restant simple à ajuster ; (2) interprétable à faible degré.
- **Limites :** (1) explosion du nombre de termes et sur-apprentissage quand $d$ ou $p$ grandit ; (2) comportement instable hors du domaine des données (extrapolation).

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_regression_polynomiale():
    g = np.random.default_rng(GRAINE)
    x = np.sort(g.uniform(-1, 1, 22)); y = 1 - 2 * x ** 2 + .5 * x + g.normal(0, .15, 22)
    xs = np.linspace(-1, 1, 300)
    fig, ax = plt.subplots(figsize=(7, 4.2))
    ax.scatter(x, y, color='k', zorder=3, label='observations')
    for d, c, nom in [(1, GRIS, 'degré 1 : sous-ajustement'), (2, VERT, 'degré 2 : adapté'), (12, ROUGE, 'degré 12 : sur-ajustement')]:
        ax.plot(xs, np.polynomial.Polynomial.fit(x, y, d)(xs), color=c, lw=2, label=nom)
    ax.set(ylim=(-2, 1.8), xlabel='$x$', ylabel='$y$', title='Régression polynomiale : le degré règle la flexibilité')
    ax.legend(fontsize=8); plt.tight_layout(); plt.show()
schema_regression_polynomiale()
```

**Comment lire le schéma.** Les trois courbes ajustent les mêmes points noirs : le degré 1 (gris) est trop rigide, le degré 2 (vert) suit la tendance, le degré 12 (rouge) colle au bruit et **oscille** aux bords. Le degré est donc un hyperparamètre de flexibilité. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (gestion / marketing) :** courbe de coût total en fonction de la quantité produite, ou réponse des ventes au prix (termes $p$ et $p^2$) pour estimer un prix optimal.
- **Cas d'usage 2 :** étalonnage d'un capteur dont la réponse est non linéaire.
- **Données et prérequis :** cible continue ; peu de variables d'entrée ; standardisation avant la création des puissances ; choix du degré par validation.
- **Cas peu approprié :** nombreuses variables avec un degré élevé (explosion des termes), ou prévision au-delà de la plage observée (les polynômes divergent).
- **Métriques adaptées :** MAE, RMSE, $R^2$ en validation croisée en fonction du degré.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données synthétiques ; `PolynomialFeatures(degree=2)` transforme 5 variables en 20 termes. **Hyperparamètres :** degré 2, `Ridge(alpha=10)`, standardisation avant la transformation.
- **Sortie obtenue :** MAE = **10,90**, plus élevée que la régression linéaire (9,35).
- **Ce que cela permet de conclure :** sur ces données, les 15 termes supplémentaires n'améliorent pas la prédiction. C'est cohérent : `make_regression` génère une relation **linéaire** ; les termes quadratiques n'apportent que de la variance, que le Ridge ne compense que partiellement.
- **Ce que cela ne permet pas de conclure :** que la régression polynomiale est moins bonne en général (elle l'est ici parce que la vérité est linéaire) ; l'écart de 1,5 sur 70 points n'est pas testé statistiquement, et ni le degré ni $\alpha$ n'ont été réglés par validation.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Arbre de régression (CART) — *Regression Tree*. **Famille :** régression supervisée non paramétrique. **Objectif :** prédire $y$ par des règles « si variable ≤ seuil alors… ».
- **Intuition et étapes :** on découpe récursivement l'espace des variables en régions rectangulaires ; dans chaque région la prédiction est la moyenne des $y$ d'apprentissage. Entrée : $X,y$ → apprentissage : à chaque nœud, choix de la variable et du seuil qui réduisent le plus l'erreur → sortie : valeur de la feuille atteinte.
- **Critère :** $\displaystyle \min_{j,s}\Big[\sum_{i\in R_G}(y_i-\bar y_G)^2+\sum_{i\in R_D}(y_i-\bar y_D)^2\Big]$, où $j$ est la variable, $s$ le seuil, $R_G,R_D$ les deux régions obtenues et $\bar y_G,\bar y_D$ leurs moyennes ; prédiction finale $\hat y(x)=\sum_m c_m\mathbf 1[x\in R_m]$, avec $c_m$ la moyenne de la région $R_m$.
- **Paramètres / hyperparamètres :** paramètres : variables, seuils et valeurs des feuilles. Hyperparamètres : `max_depth=5`, `min_samples_leaf`, `min_samples_split`.
- **Avantages :** (1) règles lisibles, pas de standardisation nécessaire ; (2) capte non-linéarités et interactions automatiquement.
- **Limites :** (1) prédiction constante par morceaux : mauvaise pour une relation linéaire lisse et aucune extrapolation ; (2) forte variance : un petit changement des données peut changer l'arbre.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_arbre_regression():
    from sklearn.tree import DecisionTreeRegressor, plot_tree
    g = np.random.default_rng(GRAINE)
    x = np.sort(g.uniform(0, 6, 60)); y = np.sin(x) + g.normal(0, .2, 60)
    arbre = DecisionTreeRegressor(max_depth=2).fit(x[:, None], y)
    xs = np.linspace(0, 6, 400)
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4.2), gridspec_kw={'width_ratios': [1, 1.3]})
    a1.scatter(x, y, s=14, color=BLEU, label='observations'); a1.plot(xs, arbre.predict(xs[:, None]), color=ORANGE, lw=2.5, label='prédiction : constante par feuille')
    for seuil in arbre.tree_.threshold[arbre.tree_.feature >= 0]: a1.axvline(seuil, color=GRIS, ls='--', lw=1)
    a1.set(xlabel='$x$', ylabel='$y$', title='Partition de l\'axe $x$ (traits pointillés = seuils)'); a1.legend(fontsize=8)
    plot_tree(arbre, feature_names=['x'], filled=True, impurity=False, ax=a2, fontsize=8, precision=2)
    a2.set_title('Arbre correspondant (profondeur 2)'); plt.tight_layout(); plt.show()
schema_arbre_regression()
```

**Comment lire le schéma.** À gauche, l'axe $x$ est découpé aux seuils (traits pointillés) ; sur chaque segment la prédiction orange est **constante**. À droite, le même modèle sous forme d'arbre : on part de la racine, on teste « $x\le$ seuil ? » et on descend jusqu'à une feuille qui donne la valeur prédite. (Données illustratives, profondeur 2.)

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / assurance) :** estimer le panier moyen ou le coût d'un sinistre par segments de clients décrits par des règles faciles à expliquer.
- **Cas d'usage 2 :** barèmes de tarification par tranches.
- **Données et prérequis :** variables numériques ou catégorielles encodées ; pas de standardisation ; profondeur ou taille minimale des feuilles à limiter pour éviter le sur-apprentissage.
- **Cas peu approprié :** relation linéaire lisse bruitée (comme ici) ou prévision de tendance hors de la plage observée : l'arbre fait des marches d'escalier et n'extrapole pas.
- **Métriques adaptées :** MAE, RMSE, $R^2$ ; profondeur et nombre de feuilles pour juger la complexité.

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données synthétiques. **Hyperparamètres :** `max_depth=5` (au plus 32 feuilles), `random_state=42`.
- **Sortie obtenue :** MAE = **37,72**, soit environ quatre fois celle de la régression linéaire (9,35).
- **Ce que cela permet de conclure :** pour une relation linéaire avec 210 points d'apprentissage, un arbre unique, constant par morceaux, approxime mal et a une forte variance.
- **Ce que cela ne permet pas de conclure :** que les arbres sont mauvais en général : la cible a été fabriquée linéaire, ce qui désavantage ce type de modèle. La profondeur 5 a été fixée sans validation ; une autre profondeur donnerait un autre résultat.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Forêt aléatoire (régression) — *Random Forest Regressor*. **Famille :** régression supervisée, méthode d'ensemble (bagging). **Objectif :** réduire la variance d'un arbre en moyennant de nombreux arbres différents.
- **Intuition et étapes :** on tire $B$ échantillons bootstrap (avec remise), on entraîne un arbre sur chacun, éventuellement en limitant à chaque nœud le nombre de variables candidates, puis on moyenne leurs prédictions. Entrée : $X,y$ → $B$ arbres → moyenne.
- **Critère :** $\displaystyle \hat y(x)=\frac1B\sum_{b=1}^{B}T_b(x)$, où $T_b$ est le $b$-ième arbre entraîné sur son échantillon bootstrap et $B$ le nombre d'arbres.
- **Paramètres / hyperparamètres :** paramètres : ceux de chaque arbre (seuils, feuilles). Hyperparamètres : `n_estimators=80` ($B$), `max_depth`, `max_features`, `min_samples_leaf` (valeurs par défaut de scikit-learn hors `n_estimators`).
- **Avantages :** (1) bien moins instable qu'un arbre unique, peu de réglages ; (2) donne une importance des variables et une erreur « out-of-bag » sans jeu de validation séparé.
- **Limites :** (1) moins interprétable qu'un arbre ; (2) n'extrapole pas et reste moins efficace qu'un modèle linéaire quand la relation est linéaire.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_foret_regression():
    from sklearn.tree import DecisionTreeRegressor
    g = np.random.default_rng(GRAINE)
    x = np.sort(g.uniform(0, 6, 70)); y = np.sin(x) + g.normal(0, .3, 70); xs = np.linspace(0, 6, 400)
    preds = []
    fig, ax = plt.subplots(figsize=(7, 4.2))
    for b in range(25):
        idx = g.integers(0, len(x), len(x))                                   # échantillon bootstrap
        p = DecisionTreeRegressor(max_depth=5, random_state=b).fit(x[idx, None], y[idx]).predict(xs[:, None]); preds.append(p)
        ax.plot(xs, p, color=BLEU, alpha=.18, lw=1, label='arbres individuels (bootstrap)' if b == 0 else None)
    ax.scatter(x, y, s=12, color='k', label='observations', zorder=3)
    ax.plot(xs, np.mean(preds, axis=0), color=ORANGE, lw=3, label='moyenne de la forêt'); ax.plot(xs, np.sin(xs), '--', color=VERT, lw=2, label='fonction vraie $\\sin x$')
    ax.set(xlabel='$x$', ylabel='$y$', title='Forêt aléatoire : la moyenne lisse la variance des arbres'); ax.legend(fontsize=8); plt.tight_layout(); plt.show()
schema_foret_regression()
```

**Comment lire le schéma.** Chaque trait bleu pâle est la prédiction d'un arbre entraîné sur un échantillon bootstrap : ils sont irréguliers et diffèrent entre eux. La courbe orange épaisse est leur **moyenne** : beaucoup plus lisse, plus proche de la fonction vraie (pointillés verts). C'est la réduction de variance du bagging. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (logistique / commerce) :** prévoir la demande d'un produit ou un délai de livraison à partir de variables mixtes (prix, saison, promotions, distance).
- **Cas d'usage 2 :** estimation de la valeur d'un bien immobilier ou d'un véhicule.
- **Données et prérequis :** données tabulaires, variables numériques ou encodées ; pas de standardisation ; données suffisamment nombreuses (centaines d'observations au moins).
- **Cas peu approprié :** prévision d'une tendance croissante au-delà de la plage d'entraînement (la forêt reste « plate » à la dernière valeur vue) ; besoin d'un modèle explicite et auditable.
- **Métriques adaptées :** MAE, RMSE, $R^2$ ; erreur out-of-bag ; importance des variables.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données synthétiques. **Hyperparamètres :** `n_estimators=80`, profondeur non limitée (défaut), `random_state=42`.
- **Sortie obtenue :** MAE = **24,37**, contre 37,72 pour un arbre unique de profondeur 5 (réduction d'environ 35 %).
- **Ce que cela permet de conclure :** moyenner des arbres améliore nettement un arbre isolé, conformément à la réduction de variance attendue.
- **Ce que cela ne permet pas de conclure :** la forêt reste bien moins précise que la régression linéaire (9,35) : la cible est linéaire par construction. Comparer un arbre de profondeur 5 et une forêt de profondeur libre mélange deux effets (moyennage et profondeur) ; le résultat est issu d'un seul découpage et n'est pas réglé.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Boosting de gradient (régression, version à histogrammes) — *Gradient Boosting / HistGradientBoostingRegressor*. **Famille :** régression supervisée, méthode d'ensemble séquentielle. **Objectif :** construire un modèle précis en ajoutant des arbres simples qui corrigent les erreurs du modèle courant.
- **Intuition et étapes :** on part d'une prédiction constante ; à chaque tour on ajuste un petit arbre sur les **résidus** (le gradient négatif de la perte) puis on l'ajoute avec un petit pas. Les variables sont discrétisées en histogrammes (jusqu'à 255 classes) pour accélérer le calcul.
- **Critère :** $F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$, où $h_m$ est un arbre ajusté sur les pseudo-résidus $r_{im}=-\partial L(y_i,F)/\partial F\big|_{F=F_{m-1}(x_i)}$, $L$ la perte (quadratique ici), $\eta$ le taux d'apprentissage et $m=1,\dots,M$ avec $M$ = `max_iter`.
- **Paramètres / hyperparamètres :** paramètres : les arbres $h_m$. Hyperparamètres : `max_iter=80` ($M$), `learning_rate` ($\eta$, 0,1 par défaut), `max_leaf_nodes`, `l2_regularization`.
- **Avantages :** (1) souvent très précis sur données tabulaires ; (2) gère nativement les valeurs manquantes et reste rapide grâce aux histogrammes.
- **Limites :** (1) sensible au sur-apprentissage si trop d'itérations ou de profondeur, et nombreux hyperparamètres à régler ; (2) boîte noire, entraînement séquentiel.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_boosting_regression():
    from sklearn.tree import DecisionTreeRegressor
    g = np.random.default_rng(GRAINE)
    x = np.sort(g.uniform(0, 6, 60)); y = np.sin(x) + .3 * x + g.normal(0, .15, 60); xs = np.linspace(0, 6, 300)
    F, Fs, eta, etapes = np.full_like(y, y.mean()), np.full_like(xs, y.mean()), .3, {0: None}
    etapes[0] = (F.copy(), Fs.copy())
    for m in range(1, 41):
        h = DecisionTreeRegressor(max_depth=2, random_state=m).fit(x[:, None], y - F)   # arbre ajusté sur les résidus
        F += eta * h.predict(x[:, None]); Fs += eta * h.predict(xs[:, None])
        if m in (3, 40): etapes[m] = (F.copy(), Fs.copy())
    fig, axes = plt.subplots(1, 3, figsize=(12, 3.6), sharey=True)
    for ax, (m, (Fm, Fsm)) in zip(axes, etapes.items()):
        ax.scatter(x, y, s=12, color='k'); ax.plot(xs, Fsm, color=ORANGE, lw=2.5); ax.vlines(x, y, Fm, color=ROUGE, lw=.8, alpha=.7)
        ax.set(title=f'Après {m} arbre(s)' if m else 'Départ : $F_0$ = moyenne', xlabel='$x$')
    axes[0].set_ylabel('$y$'); fig.suptitle('Gradient boosting : chaque arbre corrige les résidus (traits rouges) du modèle courant', y=1.02, fontsize=10)
    plt.tight_layout(); plt.show()
schema_boosting_regression()
```

**Comment lire le schéma.** Gauche : le modèle de départ est la moyenne (droite orange) ; les traits rouges sont les résidus. Au centre puis à droite, après 3 puis 40 arbres, la courbe orange épouse de mieux en mieux les points et les traits rouges raccourcissent : chaque arbre corrige ce que les précédents ont mal prédit. (Données illustratives, pas $\eta=0{,}3$.)

### Contextes d'utilisation

- **Cas d'usage 1 (commerce / logistique) :** prévoir les ventes ou les volumes à expédier à partir de nombreuses variables hétérogènes (prix, promotions, météo, calendrier).
- **Cas d'usage 2 :** estimer le temps de livraison ou le coût d'un sinistre.
- **Données et prérequis :** données tabulaires de taille moyenne à grande ; valeurs manquantes acceptées ; réglage par validation (itérations, pas, complexité).
- **Cas peu approprié :** très petits échantillons (sur-apprentissage facile) ; relation purement linéaire ou extrapolation hors domaine ; exigence forte d'explicabilité.
- **Métriques adaptées :** MAE, RMSE, $R^2$ ; courbe d'erreur de validation en fonction du nombre d'itérations.

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données synthétiques. **Hyperparamètres :** `max_iter=80`, `random_state=42` (les autres valeurs sont celles par défaut).

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Régression logistique — *Logistic Regression*. **Famille :** classification supervisée (binaire ; extension multinomiale). **Objectif :** estimer la probabilité d'appartenir à la classe 1 et décider par un seuil.
- **Intuition et étapes :** un score linéaire est transformé en probabilité par la fonction sigmoïde ; les coefficients sont appris en maximisant la vraisemblance (équivalent : minimiser l'entropie croisée, avec pénalité $L^2$ par défaut). Entrée : $X$ standardisée, $y\in\{0,1\}$ → apprentissage → sortie : probabilité puis classe (seuil 0,5).
- **Critère :** $\displaystyle P(y=1\mid x)=\sigma(\beta_0+x^\top\beta)=\frac1{1+e^{-(\beta_0+x^\top\beta)}}$ ; perte : $-\sum_i\big[y_i\ln p_i+(1-y_i)\ln(1-p_i)\big]+\frac1{2C}\lVert\beta\rVert_2^2$, avec $p_i=P(y_i=1\mid x_i)$ et $C>0$ l'inverse de la force de régularisation.
- **Paramètres / hyperparamètres :** paramètres : $\beta_0,\beta$. Hyperparamètres : $C$ (1,0 par défaut), `penalty`, `solver`, `max_iter=1000`, seuil de décision (0,5).
- **Avantages :** (1) sortie probabiliste et coefficients interprétables ($e^{\beta_j}$ = rapport de cotes) ; (2) rapide, excellent modèle de référence.
- **Limites :** (1) frontière de décision linéaire : échoue sur des structures non linéaires sans variables supplémentaires ; (2) sensible à la colinéarité et à la mise à l'échelle.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_regression_logistique():
    from sklearn.linear_model import LogisticRegression
    g = np.random.default_rng(GRAINE)
    x = np.r_[g.normal(-1, 1, 30), g.normal(1.5, 1, 30)]; y = np.r_[np.zeros(30), np.ones(30)]
    m = LogisticRegression().fit(x[:, None], y); xs = np.linspace(-4, 5, 300)
    seuil = -m.intercept_[0] / m.coef_[0, 0]
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.scatter(x, y, c=np.where(y == 1, ORANGE, BLEU), s=22, zorder=3)
    ax.plot(xs, m.predict_proba(xs[:, None])[:, 1], color='k', lw=2, label=r'$P(y=1\mid x)=\sigma(\beta_0+\beta_1x)$')
    ax.axhline(.5, color=GRIS, ls=':'); ax.axvline(seuil, color=ROUGE, ls='--', label=f'frontière de décision $x\\approx${seuil:.2f} (P = 0,5)')
    ax.set(xlabel='variable explicative $x$', ylabel='probabilité estimée', title='Régression logistique : une sigmoïde transforme le score linéaire en probabilité')
    ax.legend(fontsize=8, loc='center left'); plt.tight_layout(); plt.show()
schema_regression_logistique()
```

**Comment lire le schéma.** Les points bleus (classe 0) et orange (classe 1) sont placés selon leur $x$ ; la courbe noire est la probabilité estimée $P(y=1\mid x)$. La frontière de décision est l'abscisse où cette probabilité vaut 0,5 (trait rouge) : à gauche on prédit 0, à droite on prédit 1. (Données illustratives à une variable ; avec plusieurs variables la frontière devient un hyperplan.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance / marketing) :** scoring de crédit (probabilité de défaut) ou probabilité de désabonnement (*churn*) d'un client.
- **Cas d'usage 2 :** aide au diagnostic médical (présence ou absence d'une maladie).
- **Données et prérequis :** cible binaire ; variables numériques standardisées ou indicatrices ; pas de colinéarité forte ; assez d'observations dans chaque classe.
- **Cas peu approprié :** frontières fortement non linéaires (classes en anneaux, XOR) tant que l'on n'ajoute pas de variables d'interaction ; le modèle serait alors systématiquement trompé.
- **Métriques adaptées :** F1, précision, rappel, ROC-AUC ou PR-AUC (si classes déséquilibrées), log-loss et calibration des probabilités.

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** `Xc_ent` (300 × 6) et `yc_ent` ; test de 100 observations (découpage stratifié). Données **synthétiques** (`make_classification`). **Hyperparamètres :** `max_iter=1000`, $C=1$ par défaut, `StandardScaler` dans le `pipeline`.
- **Sortie obtenue :** F1 = **0,811** (classe positive = 1, seuil 0,5).
- **Ce que cela permet de conclure :** le modèle linéaire réalise un compromis précision/rappel correct mais inférieur à plusieurs modèles non linéaires de la série (0,89 à 0,95). Une explication plausible : `make_classification` génère par défaut **deux amas par classe**, structure qu'une frontière linéaire capte mal (hypothèse non testée ici).
- **Ce que cela ne permet pas de conclure :** F1 seul ne montre ni précision, ni rappel, ni AUC, ni matrice de confusion ; 100 observations de test signifient qu'un écart de quelques points correspond à quelques individus, sans intervalle de confiance ; le seuil de 0,5 n'est pas optimisé.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** k plus proches voisins — *k-Nearest Neighbors (k-NN)*. **Famille :** classification supervisée non paramétrique (apprentissage « paresseux »). **Objectif :** classer un point selon la classe majoritaire de ses voisins les plus proches.
- **Intuition et étapes :** « dis-moi qui sont tes voisins, je te dirai qui tu es ». Entrée : base d'apprentissage mémorisée (aucun calcul à l'apprentissage) → à la prédiction : calcul des distances au nouveau point, sélection des $k$ plus proches → vote majoritaire.
- **Critère :** $\displaystyle \hat y(x)=\arg\max_{c}\sum_{i\in N_k(x)}\mathbf 1[y_i=c]$, où $N_k(x)$ est l'ensemble des $k$ observations d'apprentissage les plus proches de $x$ pour la distance euclidienne $d(x,x')=\lVert x-x'\rVert_2$ et $c$ une classe.
- **Paramètres / hyperparamètres :** aucun paramètre appris. Hyperparamètres : $k$ (`n_neighbors=5`), métrique de distance, pondération par la distance.
- **Avantages :** (1) très simple, aucune hypothèse sur la forme de la frontière ; (2) s'adapte naturellement à des frontières non linéaires.
- **Limites :** (1) coût de prédiction élevé (distances à toutes les observations) et mémoire importante ; (2) très sensible à l'échelle des variables, aux variables inutiles et à la grande dimension.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_knn():
    g = np.random.default_rng(3)
    A = g.normal([0, 0], .9, (25, 2)); B = g.normal([2.2, 1.6], .9, (25, 2)); nouveau = np.array([1.0, .8]); k = 5
    X = np.vstack([A, B]); y = np.r_[np.zeros(25), np.ones(25)]
    d = np.linalg.norm(X - nouveau, axis=1); voisins = np.argsort(d)[:k]; votes = np.bincount(y[voisins].astype(int), minlength=2)
    fig, ax = plt.subplots(figsize=(6, 5.2))
    ax.scatter(*A.T, color=BLEU, label='classe 0'); ax.scatter(*B.T, color=ORANGE, label='classe 1')
    ax.scatter(*X[voisins].T, s=170, facecolors='none', edgecolors='k', lw=1.6, label=f'{k} plus proches voisins')
    ax.add_patch(Circle(nouveau, d[voisins].max(), fill=False, ls='--', color=GRIS))
    ax.scatter(*nouveau, marker='*', s=300, color=VERT, zorder=5, label='nouveau point')
    ax.set(aspect='equal', xlabel='variable 1 (standardisée)', ylabel='variable 2 (standardisée)', title=f'k-NN (k={k}) : vote {votes[0]} contre {votes[1]} → classe {votes.argmax()}')
    ax.legend(fontsize=8); plt.tight_layout(); plt.show()
schema_knn()
```

**Comment lire le schéma.** L'étoile verte est le point à classer ; le cercle en pointillés contient ses $k=5$ plus proches voisins (entourés de noir). On compte leurs classes (le titre indique le vote) et le point reçoit la classe majoritaire. Changer $k$ agrandit ou rétrécit le cercle. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / e-commerce) :** recommander des produits à partir des « clients les plus similaires », ou classer un nouveau client dans un segment existant.
- **Cas d'usage 2 :** reconnaissance de caractères manuscrits.
- **Données et prérequis :** variables numériques **standardisées** ; dimension modérée ; base d'apprentissage représentative et de taille raisonnable (le temps de prédiction croît avec $n$).
- **Cas peu approprié :** grande dimension (centaines de variables) où les distances perdent leur sens, ou très gros volumes avec contrainte de latence.
- **Métriques adaptées :** F1, précision, rappel, ROC-AUC (via la proportion de voisins), et $k$ choisi par validation croisée.

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données synthétiques de classification (300 d'entraînement, 100 de test). **Hyperparamètres :** `n_neighbors=5` ; la standardisation est dans le `pipeline`, donc ajustée sur l'entraînement seulement.
- **Sortie obtenue :** F1 = **0,889**, supérieur à la régression logistique (0,811).
- **Ce que cela permet de conclure :** une méthode locale non linéaire fait mieux que le modèle linéaire sur ce jeu, ce qui va dans le sens d'une frontière non linéaire.
- **Ce que cela ne permet pas de conclure :** $k=5$ est arbitraire (non réglé) ; un seul découpage, et seul F1 est affiché ; on ne peut pas affirmer que k-NN soit significativement meilleur que le modèle suivant (arbre, 0,891).

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Classifieur bayésien naïf (gaussien) — *Gaussian Naive Bayes*. **Famille :** classification supervisée probabiliste. **Objectif :** attribuer la classe de plus grande probabilité a posteriori.
- **Intuition et étapes :** on applique le théorème de Bayes en supposant les variables **indépendantes sachant la classe**, chacune suivant une loi normale par classe. Apprentissage : estimer pour chaque classe sa proportion, et la moyenne et la variance de chaque variable ; prédiction : comparer les produits de densités.
- **Critère :** $\displaystyle P(c\mid x)\propto P(c)\prod_{j=1}^{p}\mathcal N\big(x_j;\mu_{jc},\sigma_{jc}^2\big)$, où $c$ est une classe, $P(c)$ sa proportion (a priori), $\mu_{jc}$ et $\sigma_{jc}^2$ la moyenne et la variance de la variable $j$ dans la classe $c$ ; $\hat y=\arg\max_c P(c\mid x)$.
- **Paramètres / hyperparamètres :** paramètres : $P(c),\mu_{jc},\sigma_{jc}^2$ (estimés). Hyperparamètres : `var_smoothing` (stabilité numérique, $10^{-9}$ par défaut), `priors`.
- **Avantages :** (1) extrêmement rapide et peu gourmand en données ; (2) bonne référence, variantes efficaces pour le texte.
- **Limites :** (1) l'hypothèse d'indépendance est rarement vraie : probabilités mal calibrées, souvent extrêmes ; (2) hypothèse gaussienne par variable.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_naive_bayes():
    g = np.random.default_rng(GRAINE)
    a, b = g.normal(0, 1, 120), g.normal(2.2, 1.2, 60)                       # deux classes, effectifs 2/3 et 1/3
    pdf = lambda x, m, s: np.exp(-(x - m) ** 2 / (2 * s ** 2)) / (s * np.sqrt(2 * np.pi))
    xs = np.linspace(-4, 6, 500); pa, pb = len(a) / 180, len(b) / 180
    ja, jb = pa * pdf(xs, a.mean(), a.std()), pb * pdf(xs, b.mean(), b.std())
    x0 = 1.2
    fig, ax = plt.subplots(figsize=(7.2, 4.2))
    ax.plot(xs, ja, color=BLEU, lw=2, label=r'$P(C_0)\,\mathcal{N}(x;\mu_0,\sigma_0^2)$'); ax.plot(xs, jb, color=ORANGE, lw=2, label=r'$P(C_1)\,\mathcal{N}(x;\mu_1,\sigma_1^2)$')
    ax.fill_between(xs, 0, np.maximum(ja, jb), where=ja >= jb, color=BLEU, alpha=.15); ax.fill_between(xs, 0, np.maximum(ja, jb), where=jb > ja, color=ORANGE, alpha=.15)
    ax.axvline(x0, color=VERT, ls='--', label=f'point à classer $x={x0}$'); ax.set(xlabel='$x$ (une variable ; pour $p$ variables, on multiplie les densités)', ylabel='numérateur de Bayes',
        title='Naive Bayes gaussien : la classe au numérateur le plus grand l\'emporte'); ax.legend(fontsize=8); plt.tight_layout(); plt.show()
schema_naive_bayes()
```

**Comment lire le schéma.** Chaque courbe est « proportion de la classe × densité gaussienne de la variable » : c'est le numérateur de la formule de Bayes. Pour une valeur $x$ donnée (trait vert), la classe dont la courbe est la plus haute l'emporte ; les zones colorées indiquent la classe prédite en chaque $x$. Avec plusieurs variables, on **multiplie** ces densités variable par variable. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (gestion / relation client) :** classer automatiquement des réclamations ou avis de clients (variante multinomiale sur le texte) ; filtrage de spams.
- **Cas d'usage 2 :** aide au diagnostic rapide à partir de quelques mesures.
- **Données et prérequis :** variables approximativement gaussiennes par classe (version gaussienne) ou comptages (version multinomiale) ; très peu de réglages.
- **Cas peu approprié :** variables fortement corrélées ou redondantes : le produit des densités compte plusieurs fois la même information et fausse les probabilités.
- **Métriques adaptées :** F1, ROC-AUC, rappel/précision ; log-loss ou score de Brier pour évaluer la calibration.

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données de classification, sans standardisation (inutile ici). **Hyperparamètres :** valeurs par défaut.
- **Sortie obtenue :** F1 = **0,755**, la valeur la plus faible parmi les classifieurs exécutés dans ce notebook.
- **Ce que cela permet de conclure :** sur ce jeu, l'hypothèse d'indépendance conditionnelle pénalise le modèle. Une explication plausible : `make_classification` introduit des corrélations entre variables à l'intérieur de chaque amas (hypothèse non testée ici).
- **Ce que cela ne permet pas de conclure :** que Naive Bayes est mauvais en général ; il reste très compétitif en classification de texte. Un seul découpage, F1 seul : pas de calibration ni d'AUC.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Arbre de décision (CART, classification) — *Decision Tree Classifier*. **Famille :** classification supervisée. **Objectif :** prédire une classe par une suite de tests simples sur les variables.
- **Intuition et étapes :** à chaque nœud, on choisit la variable et le seuil qui séparent le mieux les classes ; on répète jusqu'à une profondeur maximale ou des feuilles pures. Prédiction : on descend l'arbre et on renvoie la classe majoritaire de la feuille.
- **Critère :** impureté de Gini $G(t)=1-\sum_{k}p_{k}^2$, où $p_k$ est la proportion de la classe $k$ dans le nœud $t$ ; la coupure retenue maximise $\Delta=G(t)-\frac{n_G}{n_t}G(t_G)-\frac{n_D}{n_t}G(t_D)$ ($n_t,n_G,n_D$ : effectifs du nœud et de ses deux fils).
- **Paramètres / hyperparamètres :** paramètres : variables et seuils de chaque nœud, classes des feuilles. Hyperparamètres : `max_depth=4`, `criterion`, `min_samples_leaf`.
- **Avantages :** (1) règles explicites, faciles à expliquer à un non-spécialiste ; (2) pas de standardisation, gère variables mixtes.
- **Limites :** (1) instable : une petite modification des données peut changer l'arbre ; (2) frontières parallèles aux axes, sur-apprentissage si l'arbre est profond.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_arbre_decision():
    from sklearn.tree import DecisionTreeClassifier, plot_tree
    from sklearn.datasets import make_moons
    X, y = make_moons(120, noise=.25, random_state=GRAINE); arbre = DecisionTreeClassifier(max_depth=2, random_state=GRAINE).fit(X, y)
    from matplotlib.colors import ListedColormap
    cm = ListedColormap(['#e58139', '#399de5'])                              # mêmes couleurs que plot_tree (classe 0 orange, classe 1 bleu)
    gx, gy, Z = grille_decision(arbre.predict, (-1.6, 2.6), (-1, 1.6))
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(11.5, 4.4), gridspec_kw={'width_ratios': [1, 1.4]})
    a1.contourf(gx, gy, Z, alpha=.2, cmap=cm); a1.scatter(*X.T, c=y, cmap=cm, edgecolor='k', s=22)
    a1.set(xlabel='$x_1$', ylabel='$x_2$', title='Régions rectangulaires (coupures parallèles aux axes)')
    plot_tree(arbre, feature_names=['x1', 'x2'], class_names=['classe 0', 'classe 1'], filled=True, impurity=True, ax=a2, fontsize=8, precision=2)
    a2.set_title('Arbre : chaque nœud teste « variable ≤ seuil »'); plt.tight_layout(); plt.show()
schema_arbre_decision()
```

**Comment lire le schéma.** À droite, chaque nœud pose une question (« $x_2\le0{,}15$ ? ») ; on va à gauche si c'est vrai, à droite sinon, jusqu'à une feuille colorée selon sa classe majoritaire (nuance = pureté). À gauche, le même modèle vu dans le plan : les coupures sont des **droites parallèles aux axes** et délimitent des rectangles. (Données illustratives, profondeur 2.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** décision d'octroi de crédit avec des règles explicables au client et au régulateur ; tri automatique de colis ou de commandes en logistique (règles de routage).
- **Cas d'usage 2 :** triage médical par questions successives.
- **Données et prérequis :** variables numériques ou catégorielles ; peu de préparation ; profondeur ou taille de feuille limitées pour éviter de mémoriser le bruit.
- **Cas peu approprié :** frontière oblique ou lissée (combinaison linéaire de variables) et petits échantillons bruités : l'arbre est instable et morcelle l'espace.
- **Métriques adaptées :** F1, précision, rappel, matrice de confusion ; profondeur et nombre de feuilles comme mesure de complexité.

```python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données de classification. **Hyperparamètres :** `max_depth=4`, `random_state=42`.
- **Sortie obtenue :** F1 = **0,891**.
- **Ce que cela permet de conclure :** un arbre peu profond, très lisible, atteint une performance voisine de k-NN (0,889) et supérieure à la régression logistique (0,811) sur ce jeu.
- **Ce que cela ne permet pas de conclure :** la profondeur 4 n'est pas choisie par validation ; l'instabilité de l'arbre n'est pas mesurée (un autre découpage donnerait peut-être un autre arbre) ; seul F1 est affiché.

## 12. Forêt aléatoire — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Forêt aléatoire (classification) — *Random Forest Classifier*. **Famille :** classification supervisée, méthode d'ensemble (bagging). **Objectif :** stabiliser les arbres de décision en combinant de nombreux arbres décorrélés.
- **Intuition et étapes :** chaque arbre est entraîné sur un échantillon bootstrap et, à chaque nœud, ne considère qu'un sous-ensemble aléatoire de variables ; la prédiction est le vote des arbres. Entrée : $X,y$ → $B$ arbres → vote majoritaire.
- **Critère :** $\displaystyle \hat y(x)=\arg\max_{c}\sum_{b=1}^{B}\mathbf 1[T_b(x)=c]$, où $T_b$ est le $b$-ième arbre et $c$ une classe (scikit-learn moyenne en pratique les probabilités des arbres).
- **Paramètres / hyperparamètres :** paramètres : ceux des arbres. Hyperparamètres : `n_estimators=80`, `max_features` (par défaut $\sqrt p$ pour la classification, soit 2 variables sur 6 à chaque coupure ici), `max_depth`, `min_samples_leaf`.
- **Avantages :** (1) bonne précision avec peu de réglages ; (2) robuste au bruit, importance des variables et erreur out-of-bag disponibles.
- **Limites :** (1) moins interprétable qu'un arbre ; (2) modèle volumineux, prédiction plus lente qu'un modèle linéaire.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_foret_classification():
    from sklearn.tree import DecisionTreeClassifier
    from sklearn.datasets import make_moons
    X, y = make_moons(150, noise=.3, random_state=GRAINE); g = np.random.default_rng(GRAINE); arbres = []
    for b in range(25):
        idx = g.integers(0, len(X), len(X)); arbres.append(DecisionTreeClassifier(max_depth=4, max_features=1, random_state=b).fit(X[idx], y[idx]))
    vote = lambda P: np.mean([a.predict(P) for a in arbres], axis=0)
    fig, axes = plt.subplots(1, 4, figsize=(15, 3.7), sharey=True)
    for i, ax in enumerate(axes):
        f = (lambda P, a=arbres[i]: a.predict(P)) if i < 3 else (lambda P: vote(P) > .5)
        gx, gy, Z = grille_decision(f, (-1.8, 2.8), (-1.3, 1.9), 150)
        ax.contourf(gx, gy, Z.astype(float), alpha=.2, cmap='coolwarm'); ax.scatter(*X.T, c=y, cmap='coolwarm', edgecolor='k', s=14)
        ax.set(title=f'Arbre n°{i + 1} (bootstrap)' if i < 3 else 'Vote majoritaire de 25 arbres', xlabel='$x_1$')
    fig.suptitle('Forêt aléatoire : des arbres instables et différents, un vote plus stable', y=1.03); plt.tight_layout(); plt.show()
schema_foret_classification()
```

**Comment lire le schéma.** Les trois premiers panneaux montrent les régions de décision de trois arbres entraînés sur des échantillons bootstrap différents (variables tirées au hasard) : leurs frontières divergent. Le dernier panneau montre le **vote** de 25 arbres : la frontière est plus régulière et suit mieux la forme des deux demi-lunes. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** détection de fraude à la carte ou prédiction de défaut de paiement sur données tabulaires hétérogènes ; prévision du désabonnement client (marketing).
- **Cas d'usage 2 :** classification d'images satellitaires à partir de bandes spectrales.
- **Données et prérequis :** données tabulaires, pas de standardisation nécessaire ; nombre d'observations suffisant ; en classes déséquilibrées, utiliser `class_weight` ou un rééchantillonnage.
- **Cas peu approprié :** quand chaque décision doit pouvoir être justifiée règle par règle, ou pour du texte très haute dimension où les modèles linéaires sont plus adaptés.
- **Métriques adaptées :** F1, rappel, précision, PR-AUC en cas de déséquilibre, ROC-AUC ; erreur out-of-bag.

```python
from sklearn.ensemble import RandomForestClassifier
modele=RandomForestClassifier(n_estimators=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

- **Sortie obtenue :** F1 = **0,918**, supérieur à l'arbre unique (0,891).

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Machine à vecteurs de support (noyau RBF) — *Support Vector Machine (SVM)*. **Famille :** classification supervisée. **Objectif :** trouver la frontière qui sépare les classes avec la **marge maximale**.
- **Intuition et étapes :** seuls les points proches de la frontière (vecteurs de support) la déterminent ; le noyau RBF projette implicitement les données dans un espace où une séparation linéaire devient possible. Entrée : $X$ standardisée, $y$ → résolution d'un problème d'optimisation convexe → sortie : classe selon le signe de $f(x)$.
- **Critère :** $\displaystyle \min_{w,b,\xi}\ \tfrac12\lVert w\rVert^2+C\sum_i\xi_i$ sous $y_i(w^\top\phi(x_i)+b)\ge1-\xi_i$, $\xi_i\ge0$ ; décision $f(x)=\sum_{i\in SV}\alpha_iy_iK(x_i,x)+b$ avec le noyau $K(x,x')=\exp(-\gamma\lVert x-x'\rVert^2)$ ; $w$ normale de l'hyperplan, $\xi_i$ variables de tolérance, $\alpha_i$ multiplicateurs de Lagrange, $SV$ ensemble des vecteurs de support.
- **Paramètres / hyperparamètres :** paramètres : $\alpha_i$, $b$. Hyperparamètres : $C$ (`C=1.0`, compromis marge / erreurs), $\gamma$ (`'scale'` par défaut), type de noyau (`'rbf'`).
- **Avantages :** (1) efficace en dimension élevée et frontières non linéaires via les noyaux ; (2) la maximisation de la marge favorise la généralisation.
- **Limites :** (1) coût d'entraînement quadratique à cubique en $n$ : difficile au-delà de quelques dizaines de milliers d'observations ; (2) pas de probabilités natives et forte sensibilité à l'échelle et aux hyperparamètres $C,\gamma$.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_svm():
    from sklearn.svm import SVC
    g = np.random.default_rng(5)
    X = np.vstack([g.normal([-1.5, -1.2], .8, (30, 2)), g.normal([1.5, 1.2], .8, (30, 2))]); y = np.r_[np.zeros(30), np.ones(30)]
    m = SVC(kernel='linear', C=1.0).fit(X, y)
    gx, gy = np.meshgrid(np.linspace(-4, 4, 300), np.linspace(-4, 4, 300)); Z = m.decision_function(np.c_[gx.ravel(), gy.ravel()]).reshape(gx.shape)
    fig, ax = plt.subplots(figsize=(6.2, 5.4))
    ax.contourf(gx, gy, Z, levels=[-99, 0, 99], colors=[BLEU, ORANGE], alpha=.12)
    ax.contour(gx, gy, Z, levels=[-1, 0, 1], colors=['k', 'k', 'k'], linestyles=['--', '-', '--'])
    ax.scatter(*X.T, c=y, cmap='coolwarm', edgecolor='k', s=24)
    ax.scatter(*m.support_vectors_.T, s=190, facecolors='none', edgecolors=VERT, lw=2, label=f'vecteurs de support ({len(m.support_vectors_)})')
    ax.plot([], [], 'k-', label='hyperplan $w^\\top x+b=0$'); ax.plot([], [], 'k--', label='marges $w^\\top x+b=\\pm1$')
    ax.set(xlim=(-4, 4), ylim=(-4, 4), aspect='equal', xlabel='$x_1$', ylabel='$x_2$', title='SVM linéaire : maximiser la marge entre les classes'); ax.legend(fontsize=8, loc='lower right')
    plt.tight_layout(); plt.show()
schema_svm()
```

**Comment lire le schéma.** La ligne noire pleine est l'hyperplan séparateur ; les deux lignes pointillées délimitent la **marge**. Les points entourés de vert sont les vecteurs de support : ce sont les seuls qui comptent pour définir la frontière (déplacer un autre point ne la change pas). (Données illustratives, noyau linéaire pour la lisibilité ; le code utilise un noyau RBF.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** prédiction de faillite d'entreprises à partir de ratios comptables sur des échantillons de taille moyenne.
- **Cas d'usage 2 :** classification de textes ou de données biologiques en haute dimension.
- **Données et prérequis :** variables numériques **standardisées** ; taille raisonnable (jusqu'à quelques dizaines de milliers de lignes) ; réglage de $C$ et $\gamma$ par validation croisée.
- **Cas peu approprié :** très gros jeux de données, ou besoin de probabilités calibrées et d'un modèle explicable (le SVM RBF est une boîte noire).
- **Métriques adaptées :** F1, rappel, précision, ROC-AUC (à partir de `decision_function`).

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données de classification, standardisées dans le `pipeline`. **Hyperparamètres :** noyau RBF, `C=1.0`, $\gamma$ par défaut.
- **Sortie obtenue :** F1 = **0,898**.
- **Ce que cela permet de conclure :** le SVM à noyau non linéaire dépasse la régression logistique (0,811) et se place au niveau des autres méthodes non linéaires simples (k-NN 0,889 ; arbre 0,891).
- **Ce que cela ne permet pas de conclure :** $C$ et $\gamma$ n'ont pas été réglés ; l'écart avec k-NN et l'arbre (moins d'un point) est trop faible pour conclure sur 100 observations de test.

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Boosting de gradient (classification, version à histogrammes) — *Gradient Boosting / HistGradientBoostingClassifier*. **Famille :** classification supervisée, méthode d'ensemble séquentielle. **Objectif :** classer en additionnant des arbres qui corrigent progressivement les erreurs de probabilité.
- **Intuition et étapes :** on part d'un score constant (log-cotes de la classe majoritaire) ; à chaque tour un petit arbre est ajusté sur le gradient de la perte logistique, puis ajouté avec un pas réduit. La probabilité finale est une sigmoïde du score cumulé.
- **Critère :** $F_M(x)=F_0+\eta\sum_{m=1}^{M}h_m(x)$ et $P(y=1\mid x)=\sigma(F_M(x))$ ; $h_m$ est l'arbre du tour $m$ ajusté sur $-\partial L/\partial F$ avec $L$ la perte logistique, $\eta$ le taux d'apprentissage.
- **Paramètres / hyperparamètres :** paramètres : les arbres $h_m$. Hyperparamètres : `max_iter=80` ($M$), `learning_rate` (0,1 par défaut), `max_leaf_nodes`, `l2_regularization`.
- **Avantages :** (1) généralement parmi les meilleurs modèles sur données tabulaires ; (2) valeurs manquantes gérées nativement, entraînement rapide grâce aux histogrammes.
- **Limites :** (1) nécessite un réglage et un contrôle du sur-apprentissage (nombre d'itérations, profondeur) ; (2) modèle peu interprétable.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_boosting_classification():
    from sklearn.ensemble import HistGradientBoostingClassifier
    from sklearn.datasets import make_moons
    X, y = make_moons(200, noise=.3, random_state=GRAINE)
    fig, axes = plt.subplots(1, 3, figsize=(12.5, 3.8), sharey=True)
    for ax, M in zip(axes, (1, 10, 100)):
        m = HistGradientBoostingClassifier(max_iter=M, learning_rate=.3, max_depth=2, min_samples_leaf=5, random_state=GRAINE).fit(X, y)
        gx, gy, Z = grille_decision(lambda P: m.predict_proba(P)[:, 1], (-1.8, 2.8), (-1.3, 1.9), 150)
        ax.contourf(gx, gy, Z, levels=np.linspace(0, 1, 11), cmap='coolwarm', alpha=.45); ax.scatter(*X.T, c=y, cmap='coolwarm', edgecolor='k', s=14)
        ax.set(title=f'{M} arbre(s) ajouté(s)', xlabel='$x_1$')
    axes[0].set_ylabel('$x_2$'); fig.suptitle('Boosting de classification : la frontière (couleur = probabilité de la classe 1) s\'affine avec les itérations', y=1.03, fontsize=10)
    plt.tight_layout(); plt.show()
schema_boosting_classification()
```

**Comment lire le schéma.** La couleur de fond est la probabilité de la classe 1 estimée par le modèle (bleu = classe 0, rouge = classe 1). Avec 1 arbre, la frontière est très grossière ; avec 10 puis 100 arbres, elle s'affine et finit par épouser le bruit (zones rouges isolées, signe de sur-apprentissage possible). Le nombre d'itérations est donc à contrôler. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance / marketing) :** scoring de crédit et de défaut, détection de clients susceptibles de partir, ciblage de campagnes.
- **Cas d'usage 2 :** détection de fraude sur transactions.
- **Données et prérequis :** données tabulaires, valeurs manquantes tolérées ; assez d'observations ; validation séparée pour régler le nombre d'itérations.
- **Cas peu approprié :** très petit échantillon (sur-apprentissage) ou exigence d'explication règle par règle (obligation réglementaire stricte).
- **Métriques adaptées :** PR-AUC et ROC-AUC, F1, rappel, log-loss.

```python
from sklearn.ensemble import HistGradientBoostingClassifier
modele=HistGradientBoostingClassifier(max_iter=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données de classification. **Hyperparamètres :** `max_iter=80`, `random_state=42`, le reste par défaut.
- **Sortie obtenue :** F1 = **0,950**, la plus élevée de toutes les cellules exécutées.
- **Ce que cela permet de conclure :** sur ce jeu, le boosting est la meilleure méthode testée (forêt 0,918 ; SVM 0,898).
- **Ce que cela ne permet pas de conclure :** que le boosting soit « le meilleur » en général : avec 100 observations de test, un écart de 3 points représente quelques individus ; un seul découpage, aucun réglage, aucune validation croisée.

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** AdaBoost (*Adaptive Boosting*). **Famille :** classification supervisée, méthode d'ensemble séquentielle. **Objectif :** combiner de nombreux classifieurs faibles (ici des souches : arbres de profondeur 1) en un classifieur fort.
- **Intuition et étapes :** à chaque tour on entraîne un classifieur faible sur les données **pondérées**, puis on augmente le poids des observations mal classées pour que le tour suivant s'y concentre ; la décision finale est un vote pondéré. Entrée : $X$, $y\in\{-1,+1\}$ → $M$ tours → vote.
- **Critère :** $\displaystyle F(x)=\operatorname{signe}\Big(\sum_{m=1}^{M}\alpha_m h_m(x)\Big)$, avec $\alpha_m=\tfrac12\ln\frac{1-\varepsilon_m}{\varepsilon_m}$ où $\varepsilon_m$ est l'erreur pondérée de $h_m$ ; mise à jour des poids $w_i\leftarrow w_i\,e^{-\alpha_m y_i h_m(x_i)}$ puis normalisation (scikit-learn utilise la variante SAMME, qui diffère de cette convention par une constante en binaire).
- **Paramètres / hyperparamètres :** paramètres : les classifieurs $h_m$, leurs poids $\alpha_m$. Hyperparamètres : `n_estimators=70` ($M$), `learning_rate` (1 par défaut), classifieur de base (souche par défaut).
- **Avantages :** (1) simple et efficace à partir de modèles très faibles ; (2) peu de réglages.
- **Limites :** (1) très sensible aux étiquettes erronées et aux valeurs aberrantes, dont le poids ne cesse de croître ; (2) entraînement séquentiel, non parallélisable.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_adaboost():
    from sklearn.tree import DecisionTreeClassifier
    g = np.random.default_rng(2)
    X = np.vstack([g.normal([-.8, -.5], 1.0, (30, 2)), g.normal([.8, .5], 1.0, (30, 2))]); y = np.r_[-np.ones(30), np.ones(30)]
    w = np.full(len(y), 1 / len(y)); souches, alphas = [], []
    fig, axes = plt.subplots(1, 4, figsize=(15.5, 3.9), sharey=True)
    for m in range(3):
        h = DecisionTreeClassifier(max_depth=1, random_state=m).fit(X, y, sample_weight=w); err = w[h.predict(X) != y].sum()
        alpha = .5 * np.log((1 - err) / err); souches.append(h); alphas.append(alpha)
        gx, gy, Z = grille_decision(h.predict, (-3.5, 3.5), (-3.5, 3.5), 120)
        ax = axes[m]; ax.contourf(gx, gy, Z, alpha=.15, cmap='coolwarm'); ax.scatter(*X.T, c=y, cmap='coolwarm', edgecolor='k', s=1500 * w + 8)
        ax.set(title=f'Tour {m + 1} : erreur pondérée = {err:.2f}, $\\alpha$ = {alpha:.2f}', xlabel='$x_1$')
        w = w * np.exp(-alpha * y * h.predict(X)); w /= w.sum()                  # les points mal classés gagnent du poids
    gx, gy, Z = grille_decision(lambda P: np.sign(sum(a * h.predict(P) for a, h in zip(alphas, souches))), (-3.5, 3.5), (-3.5, 3.5), 120)
    axes[3].contourf(gx, gy, Z, alpha=.15, cmap='coolwarm'); axes[3].scatter(*X.T, c=y, cmap='coolwarm', edgecolor='k', s=14); axes[3].set(title='Vote pondéré des 3 souches', xlabel='$x_1$')
    axes[0].set_ylabel('$x_2$'); fig.suptitle('AdaBoost : taille d\'un point = poids ; chaque souche (arbre de profondeur 1) se concentre sur les erreurs précédentes', y=1.04, fontsize=10)
    plt.tight_layout(); plt.show()
schema_adaboost()
```

**Comment lire le schéma.** La taille d'un point est proportionnelle à son poids. Au tour 1 tous les poids sont égaux ; la souche (frontière verticale) commet quelques erreurs. Ces points mal classés **grossissent** au tour 2, ce qui déplace la frontière, et ainsi de suite. Le dernier panneau montre le vote pondéré par les $\alpha_m$ des trois souches. (Données illustratives ; les erreurs et poids affichés sont calculés par le code du schéma.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance / marketing) :** scoring de clientèle à partir de règles simples (seuils sur quelques indicateurs) combinées en un score fort.
- **Cas d'usage 2 :** détection de visages en vision par ordinateur (Viola–Jones).
- **Données et prérequis :** données tabulaires numériques, étiquettes fiables ; classifieur faible simple ; réglage du nombre de tours.
- **Cas peu approprié :** données avec beaucoup de bruit d'étiquetage (par exemple enquêtes subjectives) : le poids des erreurs « incorrigibles » explose et dégrade le modèle.
- **Métriques adaptées :** F1, ROC-AUC, erreur d'entraînement et de test en fonction du nombre de tours.

```python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Entrées :** mêmes données de classification. **Hyperparamètres :** `n_estimators=70`, souches de profondeur 1 par défaut, `random_state=42`.
- **Sortie obtenue :** F1 = **0,827**.
- **Ce que cela permet de conclure :** AdaBoost avec des souches fait un peu mieux que la régression logistique (0,811) mais nettement moins bien que le boosting de gradient (0,950) et la forêt (0,918). Une explication plausible, non testée ici : des arbres de profondeur 1 peuvent capter difficilement les interactions entre variables.
- **Ce que cela ne permet pas de conclure :** que AdaBoost est inférieur au boosting de gradient en général : le classifieur de base, le pas d'apprentissage et le nombre de tours n'ont pas été réglés ; un seul découpage de 100 observations de test.

## 16. XGBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** XGBoost (*eXtreme Gradient Boosting*). **Famille :** classification (et régression) supervisée, boosting d'arbres. **Objectif :** boosting de gradient à objectif **régularisé** et optimisé au second ordre, performant et rapide.
- **Intuition et étapes :** comme le boosting de gradient, on ajoute des arbres successifs ; mais chaque arbre est construit avec les dérivées première $g_i$ et seconde $h_i$ de la perte, et la complexité des arbres est pénalisée. Entrée : $X,y$ → $M$ arbres → somme des scores → probabilité.
- **Critère :** $\displaystyle \mathcal L=\sum_i\ell(y_i,\hat y_i)+\sum_k\Omega(f_k)$, $\Omega(f)=\gamma T+\tfrac12\lambda\sum_{j}w_j^2$ ($T$ nombre de feuilles, $w_j$ poids de la feuille $j$) ; poids optimal d'une feuille $w_j^*=-\frac{G_j}{H_j+\lambda}$ où $G_j,H_j$ sont les sommes des $g_i,h_i$ dans la feuille.
- **Paramètres / hyperparamètres :** paramètres : structure et poids des arbres. Hyperparamètres : `n_estimators=80`, `max_depth=3`, `learning_rate` ($\eta$), $\lambda$ (`reg_lambda`), $\gamma$, `subsample`, `colsample_bytree`. (`eval_metric='logloss'` n'est qu'une mesure de suivi.)
- **Avantages :** (1) régularisation explicite, gestion des valeurs manquantes, parallélisation ; (2) excellentes performances sur données tabulaires.
- **Limites :** (1) nombreux hyperparamètres à régler ; (2) modèle peu interprétable et dépendance à une bibliothèque externe.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_xgboost():
    def arbre_icone(ax, x, y, couleur):
        noeuds = {'r': (x, y), 'g': (x - .55, y - .7), 'd': (x + .55, y - .7)}
        feuilles = [(x - .9, y - 1.4), (x - .2, y - 1.4), (x + .2, y - 1.4), (x + .9, y - 1.4)]
        for a, b in [('r', 'g'), ('r', 'd')]: ax.plot(*zip(noeuds[a], noeuds[b]), color=GRIS, lw=1.2)
        for f, n in zip(feuilles, ['g', 'g', 'd', 'd']): ax.plot(*zip(noeuds[n], f), color=GRIS, lw=1.2)
        for p in noeuds.values(): ax.add_patch(Circle(p, .14, fc=couleur, ec='k'))
        for p in feuilles: ax.add_patch(Rectangle((p[0] - .13, p[1] - .13), .26, .26, fc='white', ec=couleur, lw=1.5))
    fig, ax = canevas(11, 5)
    for i, (x0, c) in enumerate(zip((1.8, 5.2, 8.6), (BLEU, ORANGE, VERT))):
        arbre_icone(ax, x0, 3.7, c); ax.text(x0, 4.6, f'Arbre $f_{i + 1}$', ha='center', fontsize=10, fontweight='bold', color=c)
        ax.text(x0, 1.9, 'feuilles : $w_j^*=-\\dfrac{G_j}{H_j+\\lambda}$', ha='center', fontsize=9, color=c)
        if i < 2: ax.text(x0 + 1.7, 3.2, '+', fontsize=24, ha='center', va='center')
    boite(ax, 5.5, .95, 10, .9, 'À chaque tour : $g_i,h_i$ = dérivées 1ère et 2nde de la perte au point $i$  →  on choisit la coupure de gain maximal\n$\\mathrm{Gain}=\\frac{1}{2}\\left[\\frac{G_L^2}{H_L+\\lambda}+\\frac{G_R^2}{H_R+\\lambda}-\\frac{(G_L+G_R)^2}{H_L+H_R+\\lambda}\\right]-\\gamma$', fs=9, fc='#fdf2e9', ec=ORANGE)
    ax.text(5.5, 0.2, 'Prédiction : $\\hat y_i=\\sum_k \\eta\\,f_k(x_i)$ (η = learning rate) ; pénalité $\\Omega=\\gamma T+\\frac{1}{2}\\lambda\\sum_j w_j^2$ contre le sur-apprentissage', ha='center', fontsize=9)
    ax.set_title('XGBoost : arbres additifs, optimisés avec gradient ET hessienne, avec régularisation explicite', fontsize=10); plt.tight_layout(); plt.show()
schema_xgboost()
```

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** K-moyennes — *K-means*. **Famille :** clustering (apprentissage non supervisé). **Objectif :** partitionner les observations en $K$ groupes compacts.
- **Intuition et étapes :** (algorithme de Lloyd) 1) choisir $K$ centroïdes initiaux (initialisation k-means++) ; 2) affecter chaque point au centroïde le plus proche ; 3) recalculer chaque centroïde comme moyenne de son groupe ; répéter 2–3 jusqu'à stabilité. Entrée : $X$ (non étiqueté) → sortie : étiquette de groupe de chaque point.
- **Critère :** inertie $\displaystyle \min_{C_1..C_K,\ \mu_1..\mu_K}\ \sum_{k=1}^{K}\sum_{x_i\in C_k}\lVert x_i-\mu_k\rVert^2$, où $C_k$ est le $k$-ième groupe et $\mu_k$ son centroïde.
- **Paramètres / hyperparamètres :** paramètres : centroïdes $\mu_k$ et affectations. Hyperparamètres : $K$ (`n_clusters=3`), `n_init=10` (nombre de relances), `init`, `max_iter`.
- **Avantages :** (1) très rapide et passe à l'échelle ; (2) simple à interpréter (chaque groupe est résumé par son centroïde).
- **Limites :** (1) $K$ doit être fixé et le résultat dépend de l'initialisation (minima locaux) ; (2) suppose des groupes convexes, de tailles voisines, et est sensible aux valeurs aberrantes et à l'échelle des variables.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_kmeans():
    from sklearn.datasets import make_blobs
    X, _ = make_blobs(150, centers=[[0, 0], [5, 1], [2, 5]], cluster_std=1.1, random_state=GRAINE)
    C = X[np.random.default_rng(5).choice(len(X), 3, replace=False)].copy(); historique = [C.copy()]; coul = np.array([BLEU, ORANGE, VERT])
    aff = lambda C: ((X[:, None] - C[None]) ** 2).sum(2).argmin(1)
    for _ in range(30):
        lab = aff(C); Cn = np.array([X[lab == k].mean(0) if (lab == k).any() else C[k] for k in range(3)]); historique.append(Cn.copy())
        if np.allclose(Cn, C): break
        C = Cn
    fig, axes = plt.subplots(1, 3, figsize=(13, 4.1), sharey=True)
    titres = ['① affectation au centroïde le plus proche', '② recalcul des centroïdes (flèches)', '③ convergence : affectations stables']
    for ax, t, (Cavant, Capres) in zip(axes, titres, [(historique[0], historique[0]), (historique[0], historique[1]), (historique[-2], historique[-1])]):
        lab = aff(Cavant); ax.scatter(*X.T, c=coul[lab], s=14, alpha=.7)
        ax.scatter(*Cavant.T, marker='X', s=170, c=coul, edgecolor='k', zorder=5)
        for a, b in zip(Cavant, Capres): ax.annotate('', b, a, arrowprops=dict(arrowstyle='->', lw=2, color='k'))
        ax.set(title=t, xlabel='$x_1$'); ax.title.set_fontsize(9)
    axes[0].set_ylabel('$x_2$'); fig.suptitle('K-means (algorithme de Lloyd) : alterner affectation et mise à jour des centroïdes (✕)', y=1.03, fontsize=10); plt.tight_layout(); plt.show()
schema_kmeans()
```

**Comment lire le schéma.** Les croix sont les centroïdes, la couleur d'un point indique son groupe. Panneau ① : chaque point est colorié selon le centroïde initial le plus proche. Panneau ② : les centroïdes se déplacent (flèches) vers la moyenne de leur groupe. Panneau ③ : après quelques itérations, les affectations ne changent plus : l'algorithme a convergé (vers un minimum local de l'inertie). (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / logistique) :** segmentation de clientèle (fréquence, montant, récence) ou regroupement de points de livraison en zones de tournées.
- **Cas d'usage 2 :** compression d'image par quantification des couleurs.
- **Données et prérequis :** variables numériques **standardisées** ; distance euclidienne pertinente ; $K$ à choisir (coude, silhouette).
- **Cas peu approprié :** groupes non convexes (anneaux, croissants) ou de densités et tailles très différentes ; présence de nombreux points aberrants.
- **Métriques adaptées :** indice de silhouette, inertie (méthode du coude), indice de Davies–Bouldin, stabilité entre initialisations ; ARI ou NMI seulement si des étiquettes réelles existent (évaluation externe).

```python
from sklearn.cluster import KMeans
modele=KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(X_clu)
print('Silhouette:',silhouette_score(X_clu,modele.labels_))
plt.scatter(X_clu[:,0],X_clu[:,1],c=modele.labels_,s=12);plt.title('K-means');plt.show()
```

### Analyse de l'expérience Python

- **Entrées :** `X_clu` (240 × 2), 3 groupes gaussiens **synthétiques** (`make_blobs`, 80 points chacun). **Hyperparamètres :** `n_clusters=3`, `n_init=10`, `random_state=42`.
- **Sortie obtenue :** silhouette = **0,869** et un nuage de points colorié par groupe.
- **Ce que cela permet de conclure :** une silhouette proche de 1 indique des groupes compacts et bien séparés ; en comparant aux étiquettes de génération (`etiquettes_reference`), nous avons vérifié hors cellule que la partition obtenue leur est identique (ARI = 1).
- **Ce que cela ne permet pas de conclure :** que $K=3$ serait le bon choix sur des données réelles : ici $K$ est connu par construction et la cellule ne teste pas d'autres valeurs. La silhouette est un critère interne ; l'évaluation externe (étiquettes) n'est pas calculée dans la cellule. Les annexes (K-means manuel) retrouvent les mêmes centroïdes que scikit-learn.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Classification ascendante hiérarchique (CAH) — *Agglomerative Hierarchical Clustering*. **Famille :** clustering. **Objectif :** construire une hiérarchie de regroupements, représentée par un dendrogramme.
- **Intuition et étapes :** chaque observation commence dans son propre groupe ; à chaque étape on fusionne les deux groupes les plus proches selon un critère de lien (*linkage*), jusqu'à n'en garder qu'un. On obtient une partition en $K$ groupes en coupant le dendrogramme. Entrée : $X$ → arbre de fusions → groupes.
- **Critère (lien de Ward, défaut de scikit-learn) :** on fusionne $A$ et $B$ minimisant l'augmentation d'inertie $\displaystyle\Delta(A,B)=\frac{|A||B|}{|A|+|B|}\lVert\mu_A-\mu_B\rVert^2$, où $|A|,|B|$ sont les effectifs et $\mu_A,\mu_B$ les centroïdes.
- **Paramètres / hyperparamètres :** pas de paramètre appris au sens strict. Hyperparamètres : `n_clusters=3` (ou un seuil de distance), `linkage` (ward, complete, average, single), métrique.
- **Avantages :** (1) le dendrogramme donne une vision multi-échelle sans fixer $K$ à l'avance ; (2) déterministe, pas d'initialisation aléatoire.
- **Limites :** (1) coût en temps et mémoire d'ordre $n^2$ : inadapté aux très grands jeux ; (2) une fusion est irréversible et le résultat dépend fortement du lien choisi.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_cah():
    from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
    from sklearn.datasets import make_blobs
    X, _ = make_blobs(18, centers=3, cluster_std=.8, random_state=GRAINE); Z = linkage(X, 'ward'); lab = fcluster(Z, 3, 'maxclust')
    coupe = (Z[-3, 2] + Z[-2, 2]) / 2; coul = {1: BLEU, 2: ORANGE, 3: VERT}
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(12, 4.3), gridspec_kw={'width_ratios': [1, 1.5]})
    a1.scatter(*X.T, c=[coul[l] for l in lab], s=60, edgecolor='k')
    for i, p in enumerate(X): a1.annotate(str(i), p, (p[0] + .1, p[1] + .1), fontsize=8)
    a1.set(xlabel='$x_1$', ylabel='$x_2$', title='Observations numérotées')
    n = len(X); nc = {i: coul[lab[i]] for i in range(n)}
    for j, (u, v, _, _) in enumerate(Z): nc[n + j] = nc[int(u)] if nc[int(u)] == nc[int(v)] else GRIS   # arête colorée si tout son sous-arbre est dans un seul groupe
    dendrogram(Z, ax=a2, link_color_func=lambda k: nc[k], no_labels=False); a2.axhline(coupe, color=ROUGE, ls='--', label='coupe à 3 groupes')
    a2.set(xlabel='numéro d\'observation', ylabel='distance de fusion (Ward)', title='Dendrogramme : plus la fusion est haute, plus les groupes sont dissemblables'); a2.legend(fontsize=8)
    plt.tight_layout(); plt.show()
schema_cah()
```

**Comment lire le schéma.** À gauche, les observations numérotées et colorées selon les 3 groupes retenus. À droite, le dendrogramme : l'axe vertical est la distance de fusion ; deux feuilles se rejoignent à la hauteur où leurs groupes sont fusionnés. La ligne rouge en pointillés « coupe » l'arbre : chaque branche sous la ligne est un groupe. Plus une fusion est haute, plus les groupes réunis sont dissemblables. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (commerce international) :** regrouper des pays partenaires selon leur profil d'échanges ou des produits selon leurs caractéristiques, avec lecture hiérarchique ; taxonomie de produits en gestion de catalogue.
- **Cas d'usage 2 :** regroupement de gènes ou d'espèces (phylogénie).
- **Données et prérequis :** jeu de taille modérée (quelques milliers de lignes au plus) ; variables standardisées ; choix du lien et de la métrique.
- **Cas peu approprié :** grands volumes (>~50 000 observations) à cause du coût quadratique.
- **Métriques adaptées :** silhouette, corrélation cophénétique (fidélité du dendrogramme aux distances), choix de la coupure ; ARI si étiquettes réelles.

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

### Analyse de l'expérience Python

- **Entrées :** `X_clu` (240 × 2, synthétique). **Hyperparamètres :** `n_clusters=3`, lien de Ward et distance euclidienne par défaut.
- **Sortie obtenue :** silhouette = **0,869**, identique (à tous les chiffres affichés) à celle de K-means.
- **Ce que cela permet de conclure :** sur des groupes gaussiens bien séparés, K-means et la CAH de Ward produisent la même partition ; nous l'avons confirmé hors cellule (ARI = 1 entre les deux).
- **Ce que cela ne permet pas de conclure :** que les deux méthodes sont équivalentes en général : elles diffèrent dès que les groupes sont moins nets, de formes variées ou bruités. La cellule ne trace pas le dendrogramme et ne teste pas d'autres liens.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** DBSCAN (*Density-Based Spatial Clustering of Applications with Noise*). **Famille :** clustering par densité, avec détection de bruit. **Objectif :** trouver des groupes de forme quelconque comme des régions denses séparées par des régions peu denses.
- **Intuition et étapes :** un point est un **point cœur** s'il a au moins `min_samples` voisins (lui compris) dans un rayon $\varepsilon$ ; un groupe est formé de points cœurs voisins et des points frontières qui leur sont rattachés ; les autres points sont du **bruit** (étiquette −1). Entrée : $X$ → sortie : étiquettes (−1 = bruit).
- **Critère :** $N_\varepsilon(x)=\{x':\lVert x-x'\rVert\le\varepsilon\}$ ; $x$ est cœur si $|N_\varepsilon(x)|\ge$ `min_samples` ; deux points cœurs appartiennent au même groupe s'ils sont reliés par une chaîne de points cœurs distants d'au plus $\varepsilon$.
- **Paramètres / hyperparamètres :** aucun paramètre appris. Hyperparamètres : $\varepsilon$ (`eps=.8`), `min_samples=5`, métrique.
- **Avantages :** (1) détecte des groupes de forme arbitraire sans fixer $K$ ; (2) identifie explicitement le bruit.
- **Limites :** (1) très sensible au choix de $\varepsilon$ et incapable de gérer des densités très différentes entre groupes ; (2) perd en efficacité en grande dimension.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_dbscan():
    from sklearn.datasets import make_moons
    from sklearn.cluster import DBSCAN
    g = np.random.default_rng(GRAINE); X = np.vstack([make_moons(220, noise=.06, random_state=GRAINE)[0], g.uniform(-1.3, 2.3, (12, 2))])
    m = DBSCAN(eps=.2, min_samples=5).fit(X); lab = m.labels_; coeur = np.zeros(len(X), bool); coeur[m.core_sample_indices_] = True
    fig, ax = plt.subplots(figsize=(7, 4.6)); coul = np.array([BLEU, ORANGE, VERT, 'purple'])
    ax.scatter(*X[coeur].T, c=coul[lab[coeur] % 4], s=36, edgecolor='k', label='points cœurs')
    front = (~coeur) & (lab >= 0); ax.scatter(*X[front].T, c=coul[lab[front] % 4], s=14, label='points frontières')
    ax.scatter(*X[lab == -1].T, marker='x', color=ROUGE, s=45, label=f'bruit ({(lab == -1).sum()} points)')
    i = np.where(coeur)[0][10]; ax.add_patch(Circle(X[i], .2, fill=False, ls='--', lw=1.5)); ax.annotate('voisinage de rayon $\\varepsilon$\n(≥ min_samples points : point cœur)', X[i], (-1.3, 1.55), fontsize=8, arrowprops=dict(arrowstyle='->'))
    ax.set(aspect='equal', xlabel='$x_1$', ylabel='$x_2$', title=f'DBSCAN : {lab.max() + 1} groupes de forme libre détectés, sans fixer K'); ax.legend(fontsize=8, loc='lower right')
    plt.tight_layout(); plt.show()
schema_dbscan()
```

**Comment lire le schéma.** Les gros points sont des points cœurs, les petits des points frontières ; la croix rouge marque le bruit. Le cercle en pointillés est le voisinage de rayon $\varepsilon$ d'un point cœur : s'il contient assez de points, ce point étend son groupe à ses voisins. Ici, deux demi-lunes — que K-means séparerait mal — sont retrouvées. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (logistique) :** repérer des zones géographiques de forte densité de livraisons ou de demandes à partir de coordonnées GPS ; isoler des opérations atypiques (bruit) en gestion.
- **Cas d'usage 2 :** regroupement de positions GPS ou de trajectoires.
- **Données et prérequis :** distance pertinente et variables standardisées ; densité à peu près homogène ; réglage de $\varepsilon$ (graphe des $k$-distances).
- **Cas peu approprié :** groupes de densités très différentes, ou grande dimension où les notions de voisinage perdent leur sens.
- **Métriques adaptées :** nombre de groupes et part de bruit ; silhouette (sur les points hors bruit, avec précaution) ; DBCV ; ARI si étiquettes réelles.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

### Analyse de l'expérience Python

- **Entrées :** `X_clu` (240 × 2, synthétique). **Hyperparamètres :** `eps=.8`, `min_samples=5`.
- **Sortie obtenue :** les étiquettes présentes sont {0, 1, 2, −1}, soit **3 groupes** et **8 points de bruit** (3,3 % des 240 points).
- **Ce que cela permet de conclure :** avec ces valeurs, DBSCAN retrouve bien trois groupes. En comparant hors cellule aux étiquettes de génération, l'ARI est de 0,95 : l'écart vient des 8 points de bruit.
- **Ce que cela ne permet pas de conclure :** ces 8 points **ne sont pas de vraies anomalies** : tous ont été générés par l'un des trois groupes (ce sont des points de la périphérie, plus éloignés que $\varepsilon=0{,}8$ de leurs voisins). « Bruit » signifie « point de faible densité locale », pas « erreur de mesure » ; et la sensibilité à `eps` n'est pas explorée.

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Mélanges gaussiens — *Gaussian Mixture Models (GMM)*. **Famille :** clustering probabiliste (modèle génératif). **Objectif :** modéliser la densité des données comme une somme de lois normales et obtenir des **appartenances souples** aux groupes.
- **Intuition et étapes :** chaque groupe est une loi normale (moyenne, covariance) avec un poids. L'algorithme EM alterne : étape E (calcul des probabilités d'appartenance de chaque point) et étape M (mise à jour des poids, moyennes et covariances), jusqu'à convergence. Entrée : $X$ → paramètres du mélange → probabilités d'appartenance.
- **Critère :** $\displaystyle p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal N(x\mid\mu_k,\Sigma_k)$, avec $\pi_k\ge0$, $\sum_k\pi_k=1$ ; responsabilité $\gamma_{ik}=\dfrac{\pi_k\mathcal N(x_i\mid\mu_k,\Sigma_k)}{\sum_j\pi_j\mathcal N(x_i\mid\mu_j,\Sigma_j)}$. On maximise la log-vraisemblance.
- **Paramètres / hyperparamètres :** paramètres : $\pi_k,\mu_k,\Sigma_k$. Hyperparamètres : nombre de composantes $K$ (`n_components=3`), `covariance_type` (`'full'` par défaut), initialisation.
- **Avantages :** (1) appartenances probabilistes et groupes elliptiques de tailles et orientations différentes ; (2) critères d'information (BIC/AIC) pour choisir $K$.
- **Limites :** (1) sensible à l'initialisation (optimum local) et suppose des composantes gaussiennes ; (2) nombreux paramètres de covariance en grande dimension, risque de singularités.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_gmm():
    from sklearn.mixture import GaussianMixture
    g = np.random.default_rng(GRAINE)
    X = np.vstack([g.multivariate_normal([0, 0], [[1.4, .9], [.9, .8]], 100), g.multivariate_normal([4, 1], [[.5, 0], [0, .5]], 80), g.multivariate_normal([2, 4.5], [[.6, -.3], [-.3, 1.2]], 90)])
    m = GaussianMixture(3, random_state=GRAINE).fit(X); R = m.predict_proba(X)
    fig, ax = plt.subplots(figsize=(6.6, 5.4)); ax.scatter(*X.T, c=R, s=22, edgecolor='none')     # couleur RVB = probabilités d'appartenance
    for mu, cov in zip(m.means_, m.covariances_): ellipse_gauss(ax, mu, cov, 2, fill=False, ec='k', lw=1.8); ax.plot(*mu, 'k+', ms=14, mew=2)
    ax.set(aspect='equal', xlabel='$x_1$', ylabel='$x_2$', title='Mélange gaussien : ellipses à 2σ ; couleur = mélange des probabilités d\'appartenance')
    ax.title.set_fontsize(9); plt.tight_layout(); plt.show()
schema_gmm()
```

**Comment lire le schéma.** Les ellipses noires sont les contours à 2 écarts-types de chaque composante gaussienne (croix = moyenne) : leur forme et leur orientation sont apprises. La couleur d'un point mélange le rouge, le vert et le bleu selon ses probabilités d'appartenance : un point proche d'une ellipse est presque pur, un point à la frontière a une couleur intermédiaire. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance / marketing) :** détecter des régimes de volatilité des marchés, ou segmenter des clients avec appartenance floue (un client peut appartenir à deux segments).
- **Cas d'usage 2 :** reconnaissance de la parole (modèles acoustiques).
- **Données et prérequis :** variables continues, approximativement gaussiennes par groupe ; effectif suffisant pour estimer les covariances ; $K$ choisi par BIC/AIC.
- **Cas peu approprié :** groupes non gaussiens (croissants, anneaux) ou très peu d'observations en haute dimension.
- **Métriques adaptées :** log-vraisemblance, BIC/AIC ; silhouette (sur affectations dures), ARI si étiquettes.

```python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
```

### Analyse de l'expérience Python

- **Entrées :** `X_clu` (240 × 2, synthétique). **Hyperparamètres :** `n_components=3`, covariances complètes par défaut, `random_state=42`.
- **Sortie obtenue :** poids = **[0,333 ; 0,333 ; 0,333]** et silhouette = **0,869**.
- **Ce que cela permet de conclure :** les trois composantes ont le même poids, ce qui est cohérent avec 80 points par groupe dans `make_blobs` ; la silhouette est celle de K-means, car on la calcule sur les affectations dures (`predict`) ; nous avons vérifié hors cellule que la partition est identique (ARI = 1).
- **Ce que cela ne permet pas de conclure :** la silhouette n'exploite pas les probabilités d'appartenance, qui sont l'intérêt principal du modèle ; le nombre de composantes n'est pas testé (BIC/AIC absents) et les moyennes et covariances ne sont pas affichées.

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Analyse en composantes principales (ACP) — *Principal Component Analysis (PCA)*. **Famille :** réduction de dimension linéaire, non supervisée. **Objectif :** résumer des variables corrélées par quelques axes qui conservent le maximum de variance.
- **Intuition et étapes :** on centre (et en général on standardise) les données, on calcule la matrice de covariance, on en extrait les vecteurs propres : le premier axe est la direction de variance maximale, le deuxième est orthogonal au premier, etc. Entrée : $X$ ($n\times p$) → axes principaux → coordonnées des points sur les $q$ premiers axes.
- **Critère :** $\Sigma v_k=\lambda_kv_k$, où $\Sigma=\frac1{n-1}X_c^\top X_c$ est la matrice de covariance de $X_c$ (données centrées), $v_k$ le $k$-ième axe et $\lambda_k$ sa variance ; coordonnées $z=V_q^\top(x-\bar x)$ ; part de variance expliquée $\lambda_k/\sum_j\lambda_j$.
- **Paramètres / hyperparamètres :** paramètres : $\bar x$, vecteurs propres $v_k$. Hyperparamètre : nombre de composantes (`n_components=2`). Standardiser si les variables ont des unités différentes.
- **Avantages :** (1) déterministe, rapide, interprétable via les contributions (*loadings*) ; (2) permet visualisation, débruitage et réduction du coût.
- **Limites :** (1) purement linéaire : ne capture pas les structures courbes ; (2) maximiser la variance n'est pas maximiser la séparation entre classes, et la méthode est sensible aux valeurs aberrantes.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_acp():
    from sklearn.decomposition import PCA
    g = np.random.default_rng(GRAINE); X = g.multivariate_normal([0, 0], [[3, 2.2], [2.2, 2]], 60); p = PCA(2).fit(X); mu = X.mean(0)
    z = (X - mu) @ p.components_[0]; proj = mu + np.outer(z, p.components_[0])
    fig, ax = plt.subplots(figsize=(6.4, 5.4)); ax.scatter(*X.T, color=BLEU, s=22, label='observations')
    ax.plot(*proj.T, '.', color=ORANGE, ms=5, label='projections sur PC1'); ax.plot(np.c_[X[:, 0], proj[:, 0]].T, np.c_[X[:, 1], proj[:, 1]].T, color=GRIS, lw=.5)
    for k, c in enumerate((ROUGE, VERT)):
        v = p.components_[k] * 2 * np.sqrt(p.explained_variance_[k]); ax.annotate('', mu + v, mu, arrowprops=dict(arrowstyle='->', lw=2.5, color=c))
        ax.text(*(mu + v * 1.12), f'PC{k + 1} : {100 * p.explained_variance_ratio_[k]:.0f} % de la variance', color=c, fontsize=9, ha='center')
    ax.set(aspect='equal', xlabel='variable 1', ylabel='variable 2', title='ACP : axes orthogonaux de variance maximale'); ax.legend(fontsize=8, loc='lower right'); plt.tight_layout(); plt.show()
schema_acp()
```

**Comment lire le schéma.** Le nuage bleu est allongé selon une diagonale : la flèche rouge (PC1) pointe dans cette direction de variance maximale, la flèche verte (PC2), perpendiculaire, capte le reste. Les petits points orange sont les projections sur PC1 : on perd la distance aux traits gris, mais on garde l'essentiel de l'étalement. Le pourcentage sur chaque flèche est la part de variance expliquée. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance) :** résumer la dynamique de la courbe des taux par 2 ou 3 facteurs (niveau, pente, courbure) ; résumer des dizaines d'indicateurs macro-économiques par quelques indices (gestion).
- **Cas d'usage 2 :** prétraitement avant un modèle (réduction de colinéarité), ou cartographie de perception de marques en marketing.
- **Données et prérequis :** variables numériques **standardisées**, relations plutôt linéaires ; valeurs manquantes traitées ; pour un usage prédictif, ajustement sur l'entraînement seulement.
- **Cas peu approprié :** structures non linéaires (type « rouleau suisse »), variables catégorielles, ou lorsque l'objectif est de séparer des classes (préférer l'analyse discriminante).
- **Métriques adaptées :** part de variance expliquée (cumulée), éboulis des valeurs propres, erreur de reconstruction.

```python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
```

### Analyse de l'expérience Python

- **Entrées :** jeu réel `load_wine` (178 vins × 13 variables chimiques), standardisé. **Hyperparamètre :** `n_components=2`.
- **Sortie obtenue :** variance expliquée = **[0,362 ; 0,192]**, soit **55,4 %** cumulée (somme calculée à partir des deux valeurs affichées), et un nuage de points 2D.
- **Ce que cela permet de conclure :** les deux premiers axes conservent un peu plus de la moitié de la variance totale ; environ 44,6 % de l'information de variance est perdue dans le graphique.
- **Ce que cela ne permet pas de conclure :** les points ne sont pas coloriés par cépage (étiquettes non utilisées), donc le graphique ne dit pas si les classes sont séparables. La standardisation est ajustée sur tout le jeu : acceptable pour une description exploratoire, mais pas pour une évaluation prédictive. Le nombre de composantes n'est pas justifié (pas d'éboulis).

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** t-SNE (*t-distributed Stochastic Neighbor Embedding*). **Famille :** réduction de dimension non linéaire pour la visualisation. **Objectif :** représenter en 2D ou 3D des données de grande dimension en préservant les **voisinages locaux**.
- **Intuition et étapes :** on mesure la similarité entre points en haute dimension (probabilités $p_{ij}$ gaussiennes), on place les points au hasard en 2D avec des similarités $q_{ij}$ à queue lourde (loi de Student), puis on déplace les points par descente de gradient pour que $Q$ ressemble à $P$.
- **Critère :** $\displaystyle \min_{y_1..y_n}\ \mathrm{KL}(P\Vert Q)=\sum_{i\ne j}p_{ij}\ln\frac{p_{ij}}{q_{ij}}$, avec $p_{ij}=\frac{p_{j|i}+p_{i|j}}{2n}$, $p_{j|i}\propto\exp(-\lVert x_i-x_j\rVert^2/2\sigma_i^2)$ ($\sigma_i$ calibré par la *perplexité*) et $q_{ij}\propto(1+\lVert y_i-y_j\rVert^2)^{-1}$, $y_i$ étant la position 2D du point $i$.
- **Paramètres / hyperparamètres :** « paramètres » : les coordonnées $y_i$ elles-mêmes (il n'y a pas de modèle réutilisable). Hyperparamètres : `perplexity=20`, taux d'apprentissage, nombre d'itérations, initialisation.
- **Avantages :** (1) révèle très bien les amas locaux ; (2) souvent plus lisible qu'une ACP pour des structures non linéaires.
- **Limites :** (1) les distances entre amas, leur taille et leur nombre apparent ne sont pas interprétables, et le résultat dépend de la perplexité et de la graine ; (2) pas de projection de nouveaux points, coût élevé sur de gros jeux.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_tsne():
    from sklearn.manifold import TSNE
    from sklearn.datasets import load_digits
    d = np.linspace(0, 6, 300); chiffres = load_digits(); n = 600
    Y = TSNE(2, perplexity=30, init='pca', random_state=GRAINE).fit_transform(chiffres.data[:n])
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(12, 4.6), gridspec_kw={'width_ratios': [1, 1.3]})
    a1.plot(d, np.exp(-d ** 2 / 2), color=BLEU, lw=2, label='haute dimension : noyau gaussien'); a1.plot(d, 1 / (1 + d ** 2), color=ORANGE, lw=2, label='basse dimension : noyau de Student (1 ddl)')
    a1.set(xlabel='distance entre deux points', ylabel='similarité (non normalisée)', title='Queue lourde : les points éloignés peuvent se repousser'); a1.legend(fontsize=8)
    sc = a2.scatter(*Y.T, c=chiffres.target[:n], cmap='tab10', s=10); plt.colorbar(sc, ax=a2, label='chiffre manuscrit (étiquette, non utilisée par t-SNE)')
    a2.set(xlabel='dimension t-SNE 1', ylabel='dimension t-SNE 2', title='Exemple : 600 chiffres (64 pixels) projetés en 2D'); plt.tight_layout(); plt.show()
schema_tsne()
```

**Comment lire le schéma.** À gauche, les deux noyaux de similarité : la loi de Student (orange) décroît plus lentement que la gaussienne (bleu), ce qui permet de **repousser** davantage les points éloignés et de séparer les amas. À droite, un exemple sur 600 chiffres manuscrits (64 pixels) : les chiffres semblables forment des îlots ; les couleurs (étiquettes) ne servent qu'à lire la carte, t-SNE ne les utilise pas. Les axes n'ont pas de signification propre.

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / finance) :** explorer visuellement la structure d'une base de clients ou de transactions décrits par de nombreuses variables, pour repérer des segments ou des comportements atypiques.
- **Cas d'usage 2 :** visualiser des cellules en génomique ou des représentations apprises (embeddings) de textes ou d'images.
- **Données et prérequis :** variables numériques standardisées ; souvent précédé d'une ACP (quelques dizaines de composantes) ; taille de quelques milliers à dizaines de milliers de points.
- **Cas peu approprié :** s'en servir comme prétraitement d'un modèle prédictif, ou pour mesurer des distances et des tailles de groupes : la carte les déforme.
- **Métriques adaptées :** divergence KL finale, indice de *trustworthiness*, conservation des $k$ plus proches voisins ; stabilité selon la perplexité et la graine.

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

### Analyse de l'expérience Python

- **Entrées :** `load_wine` (178 × 13), standardisé. **Hyperparamètres :** `perplexity=20`, `random_state=42` ; les autres valeurs sont celles par défaut de scikit-learn.
- **Sortie obtenue :** uniquement un nuage de points 2D, sans valeur numérique affichée ni couleur.
- **Ce que cela permet de conclure :** la carte donne une première vue exploratoire de la structure locale des 178 vins ; elle est reproductible avec une même version de bibliothèque grâce à `random_state=42`.
- **Ce que cela ne permet pas de conclure :** aucune métrique de qualité n'est calculée ; sans étiquettes, on ne peut pas dire si les amas correspondent aux cépages ; on ne peut interpréter ni le nombre d'amas, ni leurs tailles, ni leurs distances ; la perplexité n'est pas comparée à d'autres valeurs.

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** UMAP (*Uniform Manifold Approximation and Projection*). **Famille :** réduction de dimension non linéaire (visualisation et prétraitement). **Objectif :** construire une représentation basse dimension qui conserve la structure de voisinage, avec un meilleur respect de la structure globale que t-SNE.
- **Intuition et étapes :** 1) on construit un graphe pondéré des $k$ plus proches voisins de chaque point (graphe « flou ») ; 2) on initialise une représentation basse dimension ; 3) on l'optimise pour que son propre graphe ressemble au graphe d'origine (forces d'attraction le long des arêtes, de répulsion ailleurs).
- **Critère :** poids d'une arête $\mu_{i\to j}=\exp\!\Big(-\dfrac{\max(0,\,d(x_i,x_j)-\rho_i)}{\sigma_i}\Big)$, où $\rho_i$ est la distance au plus proche voisin de $x_i$ et $\sigma_i$ une échelle locale ; symétrisation $\mu_{ij}=\mu_{i\to j}+\mu_{j\to i}-\mu_{i\to j}\mu_{j\to i}$ ; on minimise l'entropie croisée entre graphes haut et bas.
- **Paramètres / hyperparamètres :** « paramètres » : coordonnées du plongement. Hyperparamètres : `n_neighbors=15`, `min_dist` (0,1 par défaut), `n_components=2`, métrique.
- **Avantages :** (1) souvent plus rapide que t-SNE et dispose d'une méthode `transform()` pour de nouveaux points ; (2) préserve en général mieux la structure globale.
- **Limites :** (1) l'aspect de la carte dépend fortement de `n_neighbors` et `min_dist` ; (2) les distances et tailles d'amas restent d'interprétation délicate, et fixer `random_state` désactive la parallélisation.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_umap():
    from sklearn.neighbors import NearestNeighbors
    g = np.random.default_rng(GRAINE); k = 5
    X = np.vstack([g.normal([0, 0], .5, (35, 2)), g.normal([3.5, 0.5], .9, (35, 2)), g.normal([1.5, 3.5], .4, (30, 2))])
    dist, ind = NearestNeighbors(n_neighbors=k + 1).fit(X).kneighbors(X); rho = dist[:, 1]; sigma = (dist[:, 1:] - rho[:, None]).mean(1) + 1e-9
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 4.6), gridspec_kw={'width_ratios': [1.1, 1]})
    for i in range(len(X)):
        for j, dij in zip(ind[i, 1:], dist[i, 1:]):
            mu = np.exp(-max(0, dij - rho[i]) / sigma[i]); a1.plot(*X[[i, j]].T, color=BLEU, alpha=.15 + .6 * mu, lw=.6 + 1.5 * mu)
    a1.scatter(*X.T, color='k', s=14, zorder=3); a1.set(aspect='equal', xlabel='$x_1$', ylabel='$x_2$', title='① graphe flou des k voisins (épaisseur = force $\\mu_{ij}$)')
    a2.set_xlim(0, 6); a2.set_ylim(0, 5); a2.axis('off')
    etapes = ['② initialisation du plongement\n(spectrale)', '③ optimisation : attraction le long des arêtes,\nrépulsion entre non-voisins', '④ plongement basse dimension\n(+ transform() pour de nouveaux points)']
    for i, t in enumerate(etapes): boite(a2, 3, 4.2 - 1.6 * i, 5.4, 1.0, t, fs=9, fc='#fdf2e9', ec=ORANGE)
    for i in range(2): fleche(a2, (3, 3.65 - 1.6 * i), (3, 3.05 - 1.6 * i))
    fig.suptitle('UMAP : conserver la structure de voisinage d\'un graphe flou ($\\sigma_i$ approché ici ; UMAP le calibre sur $\\log_2 k$)', y=1.02, fontsize=10); plt.tight_layout(); plt.show()
schema_umap()
```

**Comment lire le schéma.** À gauche : le graphe des 5 plus proches voisins de chaque point ; plus un trait est épais et foncé, plus le lien est fort ($\mu_{ij}$ élevé). À droite : les étapes suivantes — initialisation puis optimisation du plongement (attraction le long des arêtes, répulsion entre non-voisins) — qui produisent la carte basse dimension. Dans le schéma, $\sigma_i$ est approché pour la lisibilité (UMAP le calibre sur $\log_2 k$). (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / commerce) :** cartographier un portefeuille de clients ou de produits à partir de nombreuses variables, afin de repérer des segments à explorer.
- **Cas d'usage 2 :** analyse de données de cellules uniques en biologie, ou exploration d'embeddings de documents.
- **Données et prérequis :** variables numériques standardisées ; bibliothèque `umap-learn` ; quelques milliers à millions de points.
- **Cas peu approprié :** interprétation quantitative des distances ou des tailles ; très petits échantillons où le graphe de voisinage est peu fiable.
- **Métriques adaptées :** *trustworthiness* et *continuity*, conservation des plus proches voisins, qualité d'un clustering en aval.

```python
import umap
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=umap.UMAP(n_components=2,n_neighbors=15,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('UMAP');plt.show()
```

### Analyse de l'expérience Python

- **Entrées :** `load_wine` (178 × 13) standardisé, identique à l'ACP et t-SNE.
- **Hyperparamètres :** `n_neighbors=15`, `n_components=2`, `random_state=42`.
- **Observations sur la figure :** On observe un regroupement très net en 3 groupes principaux bien distincts et resserrés. Par rapport à l'ACP qui présente un étalement continu et des chevauchements, et à t-SNE qui montre des structures plus dispersées, UMAP préserve à la fois la structure globale et les voisinages locaux de façon optimale pour ce jeu de données.

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Forêt d'isolement — *Isolation Forest*. **Famille :** détection d'anomalies non supervisée. **Objectif :** repérer les observations rares et différentes du reste.
- **Intuition et étapes :** une observation atypique est **facile à isoler** : on coupe l'espace par des seuils choisis au hasard sur des variables choisies au hasard ; un point aberrant se retrouve seul après très peu de coupures, un point normal en demande beaucoup. On construit de nombreux arbres d'isolement et on moyenne la profondeur d'isolement.
- **Critère :** score $\displaystyle s(x,n)=2^{-\,\mathbb E[h(x)]/c(n)}$, où $h(x)$ est la profondeur à laquelle $x$ est isolé dans un arbre, $\mathbb E[h(x)]$ sa moyenne sur la forêt, $n$ la taille de l'échantillon des arbres et $c(n)\approx2\ln(n-1)+1{,}154-\frac{2(n-1)}{n}$ la profondeur moyenne d'une recherche infructueuse (normalisation) ; $s$ proche de 1 : anomalie probable.
- **Paramètres / hyperparamètres :** paramètres : les coupures aléatoires de chaque arbre. Hyperparamètres : `n_estimators` (100 par défaut), `max_samples`, `contamination=.02` (proportion attendue d'anomalies, qui fixe le seuil d'alerte).
- **Avantages :** (1) rapide (complexité quasi linéaire) et sans calcul de distance ; (2) adapté aux données de dimension moyenne et sans étiquettes.
- **Limites :** (1) `contamination` impose le nombre d'alertes : on en obtient toujours une proportion fixée, même s'il n'y a pas d'anomalie ; (2) les anomalies proches d'un amas dense peuvent être « masquées » ; coupures parallèles aux axes.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_isolation_forest():
    from sklearn.datasets import make_blobs
    g = np.random.default_rng(GRAINE)
    X = np.vstack([make_blobs(100, centers=[[0, 0]], cluster_std=1.0, random_state=GRAINE)[0], [[5, 5]]]); cibles = {'point atypique': np.array([5., 5.]), 'point normal': X[0]}
    def isoler(p, rng, tracer=None):
        xmin, xmax, ymin, ymax = X[:, 0].min(), X[:, 0].max(), X[:, 1].min(), X[:, 1].max(); sub = X; n = 0
        while len(sub) > 1:
            axe = rng.integers(2); lo, hi = sub[:, axe].min(), sub[:, axe].max(); s = rng.uniform(lo, hi); n += 1
            if tracer is not None: tracer.plot([s, s] if axe == 0 else [xmin, xmax], [ymin, ymax] if axe == 0 else [s, s], color=GRIS, lw=.8)
            garder = sub[:, axe] <= s if p[axe] <= s else sub[:, axe] > s; sub = sub[garder]
            if axe == 0: xmin, xmax = (xmin, s) if p[0] <= s else (s, xmax)
            else: ymin, ymax = (ymin, s) if p[1] <= s else (s, ymax)
        return n
    fig, axes = plt.subplots(1, 2, figsize=(12, 4.8), sharey=True)
    for ax, (nom, p) in zip(axes, cibles.items()):
        ax.scatter(*X.T, color=BLEU, s=14); n = isoler(p, np.random.default_rng(11), ax); moyenne = np.mean([isoler(p, np.random.default_rng(s)) for s in range(300)])
        ax.scatter(*p, s=120, color=ROUGE, zorder=5, edgecolor='k'); ax.set(title=f'{nom} : isolé en {n} coupures (moyenne sur 300 essais : {moyenne:.1f})', xlabel='$x_1$'); ax.title.set_fontsize(9)
    axes[0].set_ylabel('$x_2$'); fig.suptitle('Isolation Forest : les anomalies sont isolées avec peu de coupures aléatoires (chemin court = score d\'anomalie élevé)', y=1.02, fontsize=10); plt.tight_layout(); plt.show()
schema_isolation_forest()
```

**Comment lire le schéma.** Chaque trait gris est une coupure aléatoire qui sépare le point rouge du reste ; on ne garde que le côté contenant ce point. À gauche, le point atypique est isolé en très peu de coupures ; à droite, un point central en demande beaucoup plus. Le titre donne aussi la moyenne sur 300 tirages aléatoires pour montrer que cette différence n'est pas un hasard. (Données illustratives.)

### Contextes d'utilisation

- **Cas d'usage 1 (finance / commerce international) :** détecter des transactions bancaires ou des écritures comptables atypiques ; repérer des déclarations douanières dont la valeur unitaire s'écarte fortement de la norme du produit.
- **Cas d'usage 2 :** maintenance prédictive à partir de mesures de capteurs.
- **Données et prérequis :** variables numériques ; peu d'anomalies ; pas besoin d'étiquettes ; nécessité d'une revue humaine des alertes.
- **Cas peu approprié :** quand on dispose de nombreuses étiquettes fiables d'anomalies (préférer un classifieur supervisé) ; anomalies contextuelles (valeur normale en général, anormale à un moment précis) ou séquentielles.
- **Métriques adaptées :** sans étiquettes : stabilité du classement des scores, relecture des alertes ; avec étiquettes fiables seulement : précision, rappel, PR-AUC.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Bandit manchot (à plusieurs bras) — *Multi-armed bandit*. **Famille :** apprentissage par renforcement (cas à un seul état). **Objectif :** maximiser le gain cumulé en choisissant répétitivement parmi $K$ actions dont les gains moyens sont inconnus (compromis exploration / exploitation).
- **Intuition et étapes :** à chaque tour, l'agent tire un bras, observe un gain et met à jour son estimation de la valeur de ce bras. Dans l'exemple, les bras sont tirés **à tour de rôle** ($t \bmod 3$) : c'est de l'exploration pure, qui sert de référence.
- **Critère :** estimation $\displaystyle\hat\mu_a(t)=\frac{S_a(t)}{N_a(t)}$, où $S_a$ est la somme des gains du bras $a$ et $N_a$ son nombre de tirages ; regret $R_T=T\mu^*-\sum_{t=1}^{T}\mu_{a_t}$, où $\mu^*$ est la valeur du meilleur bras (inconnue de l'agent).
- **Paramètres / hyperparamètres :** paramètres appris : les $\hat\mu_a$. Hyperparamètres : nombre de bras $K=3$, horizon $T=500$, stratégie de choix des bras.
- **Avantages :** (1) modèle minimal du dilemme exploration/exploitation, sans modèle de l'environnement ; (2) base de nombreuses applications (tests adaptatifs).
- **Limites :** (1) pas de notion d'état : l'action n'influence pas le futur ni ne dépend d'un contexte (voir bandits contextuels) ; (2) hypothèse de gains stationnaires.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_bandits():
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 4.4), gridspec_kw={'width_ratios': [1.05, 1]})
    a1.set_xlim(0, 7); a1.set_ylim(0, 5); a1.axis('off')
    boite(a1, 1.2, 3.9, 2.0, 1.0, 'Agent\n(estime $\\hat\\mu_a$)', fs=9); boite(a1, 5.6, 3.9, 2.4, 1.0, 'Environnement\n3 machines, $p_a$ inconnues', fc='#fdf2e9', ec=ORANGE, fs=9)
    fleche(a1, (2.2, 4.1), (4.4, 4.1), 'action $a_t$ (bras tiré)'); fleche(a1, (4.4, 3.7), (2.2, 3.7), 'gain $r_t\\in\\{0,1\\}$', decal=(0, -.4))
    boite(a1, 3.4, 1.7, 5.6, 1.2, 'Mise à jour : $N_a\\leftarrow N_a+1$,  $\\hat\\mu_a\\leftarrow \\hat\\mu_a+\\dfrac{r_t-\\hat\\mu_a}{N_a}$\nDilemme : explorer (mieux connaître) ou exploiter (gagner maintenant) ?', fs=9, fc='#eaf6ee', ec=VERT)
    fleche(a1, (1.2, 3.3), (1.8, 2.3), rad=.0); a1.set_title('Boucle agent–environnement', fontsize=10)
    g = np.random.default_rng(GRAINE); p = np.array([.2, .5, .7]); gains = np.zeros(3); n = np.zeros(3); est = []
    for t in range(500):
        a = t % 3; gains[a] += g.random() < p[a]; n[a] += 1; est.append(gains / np.maximum(n, 1))
    est = np.array(est)
    for a, c in enumerate((BLEU, ORANGE, VERT)): a2.plot(est[:, a], color=c, label=f'bras {a} : estimation $\\hat\\mu_a$'); a2.axhline(p[a], color=c, ls=':')
    a2.set(xlabel='nombre de tirages $t$', ylabel='valeur estimée', ylim=(0, 1), title='Les estimations convergent vers les $p_a$ (pointillés) — illustration'); a2.legend(fontsize=8); plt.tight_layout(); plt.show()
schema_bandits()
```

**Comment lire le schéma.** À gauche, la boucle agent–environnement : l'agent choisit un bras, reçoit un gain 0 ou 1 et met à jour sa moyenne. À droite, les estimations (courbes) convergent peu à peu vers les vraies probabilités (pointillés) : au début elles sont très instables, d'où le besoin de nombreux tirages. (Simulation illustrative avec les mêmes probabilités que le code.)

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / commerce) :** test A/B/n adaptatif de bannières, objets d'e-mail ou prix promotionnels, afin de diriger progressivement le trafic vers la meilleure variante ; choix d'un transporteur ou d'un itinéraire en logistique.
- **Cas d'usage 2 :** essais cliniques adaptatifs, recommandation de contenus.
- **Données et prérequis :** un simulateur ou un flux de décisions avec retour immédiat (clic, achat) ; peu de bras ; gains observables rapidement.
- **Cas peu approprié :** décisions séquentielles où l'action modifie l'état futur (gestion de stock, trajectoire) : il faut un apprentissage par renforcement complet.
- **Métriques adaptées :** récompense cumulée, regret, fréquence de choix du meilleur bras, **variation entre graines**.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

### Analyse de l'expérience Python

- **Entrées :** simulation (non un jeu de données réel) de 3 bras de Bernoulli de probabilités vraies [0,2 ; 0,5 ; 0,7], 500 tirages en alternance cyclique. **Hyperparamètres :** graine 42, $K=3$, $T=500$.
- **Sortie obtenue :** valeurs estimées = **[0,180 ; 0,539 ; 0,723]**.
- **Ce que cela permet de conclure :** avec environ 167 tirages par bras, les estimations restent proches des vraies valeurs (écarts de 0,02 à 0,04) et l'ordre des bras est correct.
- **Ce que cela ne permet pas de conclure :** la sortie ne dit rien de la **performance** de la stratégie : l'allocation cyclique dépense 2/3 des tirages sur les mauvais bras. Un calcul théorique (non affiché) donne un gain espéré de $(0{,}2+0{,}5+0{,}7)/3\approx0{,}47$ par tirage, contre 0,7 pour toujours tirer le meilleur bras : un regret d'environ 0,23 par tirage. Un seul tirage aléatoire (graine 42) : pas de variabilité mesurée.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Stratégie $\varepsilon$-glouton — *Epsilon-greedy*. **Famille :** apprentissage par renforcement (bandits), règle d'exploration. **Objectif :** gagner le plus possible tout en continuant à explorer les bras apparemment moins bons.
- **Intuition et étapes :** à chaque tour, avec une probabilité $\varepsilon$ on tire un bras au hasard (explorer), sinon on tire le bras de plus grande valeur estimée (exploiter). Puis on observe le gain et on met à jour les estimations.
- **Critère :** $a_t=\begin{cases}\text{bras uniforme au hasard}&\text{avec probabilité }\varepsilon\\ \arg\max_a\hat\mu_a&\text{avec probabilité }1-\varepsilon\end{cases}$, où $\hat\mu_a$ est la valeur estimée du bras $a$. Dans le code, l'estimation utilisée pour décider est $(S_a+1)/(N_a+1)$ : une **initialisation optimiste** (un succès fictif) qui pousse à essayer chaque bras au début.
- **Paramètres / hyperparamètres :** paramètres : les $\hat\mu_a$. Hyperparamètres : $\varepsilon=0{,}1$, initialisation des estimations, horizon $T=1000$ ; $\varepsilon$ peut décroître au cours du temps.
- **Avantages :** (1) extrêmement simple à implémenter ; (2) garantit une exploration permanente, donc s'adapte à un environnement qui évolue lentement.
- **Limites :** (1) exploration non dirigée : avec $\varepsilon$ constant, on continue à tirer des bras clairement mauvais, d'où un regret qui croît linéairement ; (2) $\varepsilon$ est à régler.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_epsilon_greedy():
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 4.6), gridspec_kw={'width_ratios': [1, 1.1]})
    a1.set_xlim(0, 8); a1.set_ylim(0, 6); a1.axis('off')
    boite(a1, 4, 5.3, 3.4, .8, 'début du tour $t$ : tirer $u\\sim U(0,1)$', fs=9)
    a1.add_patch(Polygon([[4, 4.6], [5.4, 3.7], [4, 2.8], [2.6, 3.7]], fc='#fff7d6', ec='#b8860b', lw=1.5)); a1.text(4, 3.7, '$u<\\varepsilon$ ?', ha='center', va='center', fontsize=10)
    boite(a1, 1.4, 1.6, 2.6, 1.2, 'EXPLORER\nbras au hasard', fc='#fdf2e9', ec=ORANGE, fs=9); boite(a1, 6.6, 1.6, 2.6, 1.2, 'EXPLOITER\nbras de plus grande\nvaleur estimée', fc='#eaf6ee', ec=VERT, fs=9)
    fleche(a1, (4, 4.9), (4, 4.6)); fleche(a1, (2.9, 3.4), (1.9, 2.25), 'oui (prob. $\\varepsilon$)', decal=(-.9, 0)); fleche(a1, (5.1, 3.4), (6.1, 2.25), 'non (prob. $1-\\varepsilon$)', decal=(.9, 0))
    a1.text(4, .35, 'puis : observer le gain, mettre à jour $\\hat\\mu_a$, tour suivant', ha='center', fontsize=9); a1.set_title('Règle de décision $\\varepsilon$-greedy', fontsize=10)
    g = np.random.default_rng(GRAINE); p = np.array([.2, .5, .7]); S, N = np.zeros(3), np.zeros(3); part = []
    for t in range(1000):
        a = int(g.integers(3)) if g.random() < .1 else int(np.argmax((S + 1) / (N + 1))); S[a] += g.random() < p[a]; N[a] += 1; part.append(N / N.sum())
    part = np.array(part)
    for a, c in enumerate((BLEU, ORANGE, VERT)): a2.plot(part[:, a], color=c, lw=2, label=f'bras {a} ($p={p[a]}$)')
    a2.set(xlabel='nombre de tours $t$', ylabel='part cumulée des tirages', ylim=(0, 1), title='Illustration ($\\varepsilon=0{,}1$) : le meilleur bras finit par dominer'); a2.legend(fontsize=8); plt.tight_layout(); plt.show()
schema_epsilon_greedy()
```

**Comment lire le schéma.** À gauche, l'arbre de décision d'un tour : on tire $u$ uniformément entre 0 et 1 ; si $u<\varepsilon$ on explore, sinon on exploite. À droite, la part cumulée de tirages de chaque bras au cours du temps : le bras de probabilité 0,7 (vert) finit par concentrer l'essentiel des tirages, les autres conservant une petite part due à l'exploration. (Simulation illustrative reprenant les paramètres du code.)

### Contextes d'utilisation

- **Cas d'usage 1 (marketing / e-commerce) :** sélectionner dynamiquement la publicité ou la remise la plus rentable parmi quelques variantes, en continuant à tester les autres.
- **Cas d'usage 2 :** recommandation de contenus ou de produits.
- **Données et prérequis :** retour d'information rapide pour chaque action ; peu d'actions ; choix de $\varepsilon$ (éventuellement décroissant).
- **Cas peu approprié :** environnement très non stationnaire sans mécanisme d'oubli, ou situations où une action aléatoire est coûteuse ou dangereuse (soins médicaux, décisions financières réelles).
- **Métriques adaptées :** récompense cumulée, regret, part de tirages du meilleur bras, **variation entre graines**.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Analyse de l'expérience Python

- **Entrées :** simulation de 3 bras de Bernoulli [0,2 ; 0,5 ; 0,7], 1 000 tirages. **Hyperparamètres :** $\varepsilon=0{,}1$, estimations initiales optimistes, graine 42.
- **Sortie obtenue :** tirages = **[25 ; 49 ; 926]** ; valeurs estimées = **[0,200 ; 0,673 ; 0,693]**.
- **Ce que cela permet de conclure :** le meilleur bras (indice 2) a reçu 92,6 % des tirages : la stratégie l'a identifié et exploité. À partir des valeurs affichées, le gain total vaut $0{,}2\times25+0{,}673\times49+0{,}693\times926\approx680$, soit **0,68 par tirage**, bien au-dessus de l'allocation cyclique (≈ 0,47 en espérance).
- **Ce que cela ne permet pas de conclure :** le bras 1 est surestimé (0,673 pour une vraie valeur de 0,5) car il n'a été tiré que 49 fois : l'estimation d'un bras peu tiré est bruitée. La sortie ne donne pas le regret ni la variabilité : un seul tirage aléatoire ; il faudrait répéter sur plusieurs graines et comparer à d'autres valeurs de $\varepsilon$.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Q-learning — *Q-learning (tabulaire)*. **Famille :** apprentissage par renforcement, méthode de valeur, par différence temporelle, **hors politique** (*off-policy*). **Objectif :** apprendre une politique qui maximise la récompense cumulée actualisée dans un environnement à états discrets.
- **Intuition et étapes :** l'agent interagit avec l'environnement (état $s$ → action $a$ → récompense $r$ et état suivant $s'$) et met à jour une table $Q(s,a)$ qui estime la valeur de chaque action. Il explore avec une stratégie $\varepsilon$-gloutonne ($\varepsilon$ décroît de 1 à 0,05), mais apprend la valeur de la **meilleure** action suivante.
- **Critère :** $\displaystyle Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\big]$, où $\alpha$ est le taux d'apprentissage, $\gamma$ le facteur d'actualisation et $\max_{a'}Q(s',a')$ la meilleure valeur estimée à l'état suivant (nulle si $s'$ est terminal).
- **Paramètres / hyperparamètres :** paramètres : la table $Q$ (16 × 4). Hyperparamètres : $\alpha=0{,}3$, $\gamma=0{,}95$, $\varepsilon$ (calendrier $\max(0{,}05;\,1-\text{épisode}/2500)$), 4 000 épisodes, 100 pas maximum.
- **Avantages :** (1) simple, sans modèle de transition, converge vers la politique optimale sous conditions ; (2) peut apprendre la politique optimale en suivant une autre politique d'exploration.
- **Limites :** (1) la table devient inutilisable si l'espace d'états est grand ou continu (d'où le DQN) ; (2) apprentissage lent, sensible aux hyperparamètres, et tendance à surestimer les valeurs.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_q_learning():
    plan = ['SFFF', 'FHFH', 'FFFH', 'HFFG']; mouv = {0: (0, -1), 1: (1, 0), 2: (0, 1), 3: (-1, 0)}   # gauche, bas, droite, haut
    from collections import deque
    file, pere = deque([0]), {0: None}
    while file:                                                                 # parcours en largeur : plus court chemin S -> G
        s = file.popleft(); r, c = divmod(s, 4)
        for a, (dr, dc) in mouv.items():
            r2, c2 = r + dr, c + dc
            if 0 <= r2 < 4 and 0 <= c2 < 4 and plan[r2][c2] != 'H' and 4 * r2 + c2 not in pere: pere[4 * r2 + c2] = s; file.append(4 * r2 + c2)
    chemin, s = [], 15
    while s is not None: chemin.append(s); s = pere[s]
    chemin = chemin[::-1]
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(12.5, 4.8), gridspec_kw={'width_ratios': [1, 1.2]})
    couleurs = {'S': '#cfe8ff', 'F': '#f4faff', 'H': '#7f8c8d', 'G': '#bde5c8'}
    for s in range(16):
        r, c = divmod(s, 4); a1.add_patch(Rectangle((c, 3 - r), 1, 1, fc=couleurs[plan[r][c]], ec='k')); a1.text(c + .06, 3 - r + .86, f'{plan[r][c]} · état {s}', ha='left', va='center', fontsize=8)
    for s, t in zip(chemin[:-1], chemin[1:]):
        (r, c), (r2, c2) = divmod(s, 4), divmod(t, 4); a1.annotate('', (c2 + .5, 3 - r2 + .5), (c + .5, 3 - r + .5), arrowprops=dict(arrowstyle='->', lw=2.5, color=ROUGE))
    a1.set(xlim=(0, 4), ylim=(0, 4), aspect='equal', title=f'FrozenLake 4×4 : un plus court chemin existe ({len(chemin) - 1} pas)\nH = trou (fin, r=0), G = but (fin, r=1)'); a1.axis('off'); a1.title.set_fontsize(9)
    a2.axis('off'); a2.set_xlim(0, 10); a2.set_ylim(0, 6)
    q, r_, gam, qmax, alp = .40, 0., .95, .80, .3; cible = r_ + gam * qmax; nouveau = q + alp * (cible - q)
    boite(a2, 5, 5.0, 9, 1.0, 'Transition observée : $(s,a,r,s\')$ — valeurs d\'illustration :\n$Q(s,a)=0{,}40$ ; $r=0$ ; $\\max_{a\'}Q(s\',a\')=0{,}80$ ; $\\alpha=0{,}3$ ; $\\gamma=0{,}95$', fs=9)
    boite(a2, 5, 3.3, 9, 1.0, f'cible = r + γ·max Q(s′,·) = 0 + 0,95 × 0,80 = {cible:.2f}'.replace('.', ','), fs=10, fc='#fdf2e9', ec=ORANGE)
    boite(a2, 5, 1.6, 9, 1.0, '$Q(s,a)\\leftarrow Q+\\alpha\\,(\\mathrm{cible}-Q)=0{,}40+0{,}3\\times(' + f'{cible:.2f}'.replace('.', ',') + '-0{,}40)=' + f'{nouveau:.3f}'.replace('.', ',') + '$', fs=10, fc='#eaf6ee', ec=VERT)
    fleche(a2, (5, 4.5), (5, 3.8)); fleche(a2, (5, 2.8), (5, 2.1)); a2.set_title('Une mise à jour de Q-learning (hors politique : max sur $a\'$)', fontsize=10); plt.tight_layout(); plt.show()
schema_q_learning()
```

**Comment lire le schéma.** À gauche, la grille FrozenLake 4 × 4 : S = départ, F = case gelée sûre, H = trou (l'épisode s'arrête sans récompense), G = but (récompense 1). Les flèches rouges montrent un plus court chemin (6 pas), calculé par parcours en largeur : c'est ce qu'une bonne politique doit retrouver. À droite, un exemple chiffré d'une mise à jour de $Q$ (valeurs d'illustration).

### Contextes d'utilisation

- **Cas d'usage 1 (logistique / gestion) :** déplacement de robots dans un entrepôt à grille discrète, ou politique de réapprovisionnement de stock avec des niveaux de stock discrétisés (en simulation).
- **Cas d'usage 2 :** jeux simples, contrôle de petits systèmes.
- **Données et prérequis :** un **environnement simulé** avec états et actions discrets en petit nombre, et la possibilité de répéter de nombreux épisodes (ce n'est pas un jeu de données statique).
- **Cas peu approprié :** états continus ou de grande dimension (images) ; situations où l'exploration en conditions réelles est coûteuse ou dangereuse (marchés financiers réels sans simulateur) ; apprentissage à partir d'un historique figé.
- **Métriques adaptées :** récompense cumulée par épisode, taux de réussite, nombre de pas, courbe d'apprentissage, **variation entre graines**.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** SARSA (*State–Action–Reward–State–Action*). **Famille :** apprentissage par renforcement, méthode de valeur, par différence temporelle, **sur politique** (*on-policy*). **Objectif :** apprendre la valeur de la politique effectivement suivie (exploration comprise) et l'améliorer.
- **Intuition et étapes :** à chaque pas, l'agent est dans l'état $s$, choisit $a$, reçoit $r$, arrive en $s'$ et choisit **déjà** l'action suivante $a'$ avec la même stratégie $\varepsilon$-gloutonne ; la mise à jour utilise ce $a'$ réellement choisi, d'où le nom.
- **Critère :** $\displaystyle Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\,Q(s',a')-Q(s,a)\big]$, où $a'$ est l'action effectivement choisie en $s'$ (alors que Q-learning utilise $\max_{a'}Q(s',a')$).
- **Paramètres / hyperparamètres :** paramètres : la table $Q$ (16 × 4). Hyperparamètres : $\alpha=0{,}3$, $\gamma=0{,}95$, calendrier de $\varepsilon$ (de 1 à 0,05), 4 000 épisodes, 100 pas maximum.
- **Avantages :** (1) prend en compte le risque de l'exploration : politique souvent plus prudente près de zones dangereuses ; (2) apprentissage généralement stable.
- **Limites :** (1) converge vers la politique optimale seulement si $\varepsilon$ tend vers 0 ; (2) mêmes limites que Q-learning : table inutilisable pour de grands espaces d'états, sensibilité aux hyperparamètres.

**Comment lire le schéma.** Les deux lignes montrent le même enchaînement $s_t\to a_t\to r_{t+1}\to s_{t+1}$ ; les quatre cercles sont les valeurs $Q(s_{t+1},\cdot)$ (valeurs d'illustration). **Q-learning** (haut) prend la meilleure (cercle vert = 0,8) ; **SARSA** (bas) prend la valeur de l'action réellement tirée, ici une action exploratoire (0,2). La cible de SARSA tient donc compte du risque lié à l'exploration.

### Contextes d'utilisation

- **Cas d'usage 1 (logistique) :** robot d'entrepôt devant apprendre en conditions réelles tout en évitant les collisions : on veut une politique qui tient compte des erreurs d'exploration ; gestion de stock où une exploration trop agressive a un coût.
- **Cas d'usage 2 :** navigation de robots près d'obstacles.
- **Données et prérequis :** environnement simulé ou réel à états et actions discrets, interactions répétées ; calendrier de décroissance de $\varepsilon$.
- **Cas peu approprié :** grands espaces d'états, ou quand on veut apprendre la politique optimale à partir de données collectées par une autre politique (préférer une méthode hors politique).
- **Métriques adaptées :** récompense cumulée, taux de réussite, nombre de pas, **variation entre graines**, comparaison avec Q-learning sur le même environnement.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

- **Nom :** Réseau Q profond — *Deep Q-Network (DQN)*. **Famille :** apprentissage par renforcement profond, méthode de valeur. **Objectif :** généraliser le Q-learning à des espaces d'états grands ou continus en approchant $Q(s,a)$ par un réseau de neurones.
- **Intuition et étapes :** l'agent stocke ses transitions $(s,a,r,s',d)$ dans une **mémoire de rejeu** ; à chaque pas d'apprentissage, il tire un mini-lot aléatoire, calcule la cible de Bellman avec un **réseau cible** gelé, et met à jour le réseau en ligne par descente de gradient ; le réseau cible est recopié périodiquement.
- **Critère :** cible $y=r+\gamma\,(1-d)\max_{a'}Q(s',a';\theta^-)$ et perte $L(\theta)=\mathbb E\big[(y-Q(s,a;\theta))^2\big]$, où $\theta$ sont les poids du réseau en ligne, $\theta^-$ ceux du réseau cible, $d\in\{0,1\}$ indique un état terminal et $\gamma$ le facteur d'actualisation.
- **Paramètres / hyperparamètres :** paramètres : poids $\theta$ du réseau. Hyperparamètres : $\gamma$ (0,95 dans le code), taux d'apprentissage, taille du mini-lot, taille de la mémoire, fréquence de copie de $\theta^-$, calendrier de $\varepsilon$.
- **Avantages :** (1) traite des entrées de grande dimension (images, nombreuses variables) ; (2) le rejeu d'expérience et le réseau cible stabilisent l'apprentissage.
- **Limites :** (1) coûteux en interactions et instable, très sensible aux hyperparamètres ; (2) tendance à surestimer les valeurs et limité aux **actions discrètes**.

### Schéma explicatif

```python
# Schéma pédagogique (Matplotlib) — données illustratives, sans lien avec les données réelles de la partie B
def schema_dqn():
    fig, ax = canevas(12, 6.4)
    boite(ax, 1.5, 5.2, 2.4, 1.0, 'Environnement', fc='#fdf2e9', ec=ORANGE); boite(ax, 5.0, 5.2, 3.0, 1.0, 'Mémoire de rejeu $\\mathcal{D}$\n$(s,a,r,s\',d)$')
    boite(ax, 9.3, 5.2, 3.0, 1.0, 'Mini-lot tiré au hasard\n(décorrèle les données)')
    boite(ax, 3.0, 2.9, 3.6, 1.2, 'Réseau en ligne\n$Q(s,a;\\theta)$', fc='#eaf6ee', ec=VERT); boite(ax, 9.0, 2.9, 3.8, 1.2, 'Réseau cible (gelé)\n$Q(s\',a\';\\theta^-)$', fc='#eaf6ee', ec=VERT)
    boite(ax, 6.0, .9, 9.4, 1.0, 'cible $y=r+\\gamma(1-d)\\max_{a\'}Q(s\',a\';\\theta^-)$    perte $L(\\theta)=\\mathbb{E}\\left[(y-Q(s,a;\\theta))^2\\right]$  →  gradient → mise à jour de $\\theta$', fs=9, fc='#fdf2e9', ec=ORANGE)
    fleche(ax, (2.7, 5.2), (3.5, 5.2), ''); fleche(ax, (6.5, 5.2), (7.8, 5.2)); fleche(ax, (9.3, 4.7), (9.0, 3.5), 'états suivants $s\'$'); fleche(ax, (8.4, 4.7), (4.4, 3.5), 'états $s$, actions $a$', rad=.0, decal=(-.5, .15))
    fleche(ax, (3.0, 2.3), (4.5, 1.4)); fleche(ax, (9.0, 2.3), (7.5, 1.4)); fleche(ax, (1.8, 2.9), (1.0, 4.6), 'action $\\varepsilon$-greedy', rad=-.25, decal=(-.9, -.4))
    fleche(ax, (7.0, 3.0), (4.8, 3.0), 'copie périodique $\\theta^-\\leftarrow\\theta$', style='--', couleur=ROUGE, decal=(.2, .1))
    ax.set_title('DQN : Q-learning où la table Q est remplacée par un réseau de neurones, stabilisé par rejeu d\'expérience et réseau cible', fontsize=10); plt.tight_layout(); plt.show()
schema_dqn()
```

**Comment lire le schéma.** Suivez les flèches : l'environnement fournit des transitions stockées dans la mémoire ; un mini-lot aléatoire sert à calculer, d'un côté $Q(s,a;\theta)$ avec le réseau en ligne, de l'autre la cible $y$ avec le réseau cible gelé ; la perte entre les deux met à jour $\theta$. La flèche rouge en pointillés indique la copie périodique $\theta^-\leftarrow\theta$ ; la flèche vers l'environnement est le choix d'action $\varepsilon$-glouton.

### Contextes d'utilisation

- **Cas d'usage 1 (logistique / finance) :** pilotage d'un système complexe en simulation (gestion d'entrepôt, allocation de ressources) ; recherche de stratégies d'exécution d'ordres sur un simulateur de marché — en ayant conscience du risque de transfert vers la réalité.
- **Cas d'usage 2 :** jeux Atari (Mnih et al., 2015), contrôle robotique à actions discrètes.
- **Données et prérequis :** un simulateur rapide et un très grand nombre d'interactions ; GPU souvent utile ; réglage soigneux.
- **Cas peu approprié :** actions continues (préférer DDPG, SAC) ; budget d'interactions réduit ; absence de simulateur et exploration réelle dangereuse.
- **Métriques adaptées :** récompense cumulée moyenne par épisode, taux de réussite, courbe d'apprentissage, **variation entre graines**.

### Analyse de l'expérience Python

- **Entrées :** trois transitions fictives : récompenses [0 ; 1 ; −1], indicateurs de fin d'épisode [faux ; vrai ; faux], meilleures valeurs suivantes $\max_{a'}Q$ = [2 ; 3 ; 1], $\gamma=0{,}95$. **Aucun réseau, aucune mémoire de rejeu** : c'est une illustration **conceptuelle** du calcul de la cible.

- **Ce que cela ne permet pas de conclure :** aucun apprentissage n'a lieu : on ne peut juger ni la performance ni la stabilité d'un DQN. Une expérience DQN complète sur Gymnasium est **facultative** (partie B) et devra être signalée comme telle.
